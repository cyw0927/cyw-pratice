## Q03. RGB 값은 왜 보통 0부터 255까지인가?

### 답
각 색 성분을 8비트 정수로 표현하는 경우 0부터 255까지 총 256개의 값을 사용할 수 있기 때문이다.

### 상세 설명
비트(bit)는 0 또는 1 한 자리를 뜻한다. 8비트를 사용하면 만들 수 있는 조합은 `2**8 = 256`개다. 정수 값을 0부터 시작하므로 사용할 수 있는 범위는 0, 1, 2, ... , 255가 된다. RGB에서는 이 숫자를 각 색 성분의 세기로 사용한다. 0은 해당 색 빛이 없는 상태에 가깝고 255는 가장 강한 상태라고 생각하면 된다.

이 문제의 밝기 공식은 세 RGB 값을 평균내므로, 정상 입력만 들어온다면 밝기 역시 0부터 255 범위를 벗어나지 않는다. 이 사실을 알고 있으면 테스트 중 밝기가 300처럼 나온 경우 계산이나 입력 처리에 문제가 있다는 것을 빠르게 의심할 수 있다.

### 왜 필요한가?
입력 데이터의 정상 범위와 계산 결과의 정상 범위를 예측할 수 있다.

### 헷갈리기 쉬운 점
256단계라고 해서 최댓값이 256인 것은 아니다. 0을 포함하기 때문에 최댓값은 255다.

### 직접 해보기
`2**8`과 `2**8 - 1`을 출력해 둘의 차이를 확인한다.

```python
print('8비트 조합 개수:', 2**8)
print('0부터 시작할 때 최댓값:', 2**8 - 1)
```

## Q04. 한 픽셀 문자열 `255,0,0`은 실제로 어떤 구조인가?

입력 순간에는 문자열 하나지만 의미상으로는 R=255, G=0, B=0이라는 세 정수 값을 가진 픽셀 하나다.

이 문제에는 두 종류의 구분자가 등장한다. 픽셀과 픽셀 사이는 공백이고, 한 픽셀 내부의 R,G,B 사이는 쉼표다. 그래서 입력 처리도 두 단계로 해야 한다. 먼저 공백으로 픽셀을 나누고, 그 다음 각 픽셀을 쉼표로 나눈다.

### 헷갈리기 쉬운 점
`255,0,0 0,255,0` 전체를 바로 쉼표로만 나누면 픽셀 경계가 깨질 수 있다.

### 직접 해보기

## Q05. 검은색과 흰색의 RGB와 밝기는 각각 얼마인가?

### 답
검은색은 `(0,0,0)`이고 밝기는 0, 흰색은 `(255,255,255)`이고 밝기는 255다.

### 상세 설명
이 문제의 밝기 공식은 `(R + G + B) // 3`이다. 검은색을 넣으면 `(0+0+0)//3 = 0`, 흰색을 넣으면 `(255+255+255)//3 = 255`가 된다. 즉 RGB의 양 끝값을 넣었을 때 밝기도 자연스럽게 0과 255가 된다.

이 두 값은 단순한 예시가 아니라 테스트할 때 중요한 경계값이다. 밝기 함수를 만들었다면 가장 먼저 검정과 흰색을 넣어 계산 결과가 각각 0과 255인지 확인할 수 있다. 이런 극단값 테스트는 함수가 기본적으로 정상 동작하는지 빠르게 검증하는 데 도움이 된다.

### 왜 필요한가?
밝기 값의 가능한 최소·최대 범위를 이해하고 함수 테스트 기준을 세울 수 있다.

### 헷갈리기 쉬운 점
흰색의 RGB 합은 765지만 밝기는 765가 아니라 세 값의 평균인 255다.

### 직접 해보기
두 RGB 값을 직접 공식에 넣어 계산하고 중간 합도 출력해 본다.

```python
black=(0,0,0)
white=(255,255,255)
for name, rgb in [('black',black),('white',white)]:
    total=sum(rgb)
    brightness=total//3
    print(name, '합=', total, '밝기=', brightness)
```

## Q06. 왜 RGB 세 값을 밝기 하나로 바꾸는가?

### 답
흑백으로 분류할 때는 색의 종류보다 픽셀이 얼마나 밝고 어두운지를 하나의 숫자로 비교하는 것이 훨씬 단순하기 때문이다.

### 상세 설명
RGB를 그대로 사용하면 픽셀 하나를 비교할 때마다 세 숫자를 동시에 봐야 한다. 예를 들어 `(255,0,0)`과 `(85,85,85)` 중 어느 쪽을 더 밝다고 볼지 세 성분만 보고 바로 결정하기 어렵다. 하지만 문제에서 정한 밝기 공식으로 변환하면 두 픽셀 모두 85처럼 하나의 숫자로 표현할 수 있다. 이후에는 모든 픽셀을 같은 threshold와 비교하면 되므로 조건문이 매우 단순해진다.

중요한 점은 이 변환 과정에서 색 정보 일부가 사라진다는 것이다. 빨강, 초록, 파랑이 모두 밝기 85가 될 수 있다. 하지만 이 문제의 목표는 색을 보존하는 것이 아니라 밝고 어두운 정도로 0과 1을 나누는 것이므로 이런 단순화가 허용된다.

### 왜 필요한가?
threshold 하나로 모든 픽셀을 비교하려면 각 픽셀도 비교 가능한 대표 숫자 하나를 가져야 한다.

### 헷갈리기 쉬운 점
서로 다른 색이 같은 brightness 값을 가질 수 있다.

### 직접 해보기
빨강 `(255,0,0)`과 회색 `(85,85,85)`을 계산해 같은 값이 나오는지 확인한다.

```python
for rgb in [(255,0,0),(85,85,85),(0,255,0),(0,0,255)]:
    print(rgb, '→', sum(rgb)//3)
```

## Q07. 이 문제에서 밝기는 정확히 어떻게 계산하는가?

### 답
`brightness = (R + G + B) // 3`으로 계산한다.

### 상세 설명
먼저 R, G, B 세 값을 모두 더하고 그 합을 3으로 나눈다. 세 색 성분을 같은 비중으로 취급하는 단순 평균 방식이다. 예를 들어 `(120,120,120)`은 합이 360이고 `360//3 = 120`이므로 밝기 120이다. `(255,0,0)`은 합이 255이고 `255//3 = 85`가 된다.

실제 이미지 처리에서는 사람 눈의 민감도를 반영해서 R,G,B에 서로 다른 가중치를 곱하는 공식도 사용한다. 하지만 알고리즘 문제에서는 문제에서 지정한 공식을 그대로 따라야 한다. 더 현실적인 공식이 있다고 해서 임의로 바꾸면 예제 결과와 달라지고 오답이 된다.

### 왜 필요한가?
모든 이후 계산이 brightness 값을 기준으로 이루어지므로 가장 핵심적인 계산식이다.

### 헷갈리기 쉬운 점
실제 이미지 처리 지식을 섞어 다른 공식으로 바꾸지 않는다.

### 직접 해보기
서로 다른 RGB 세 개를 골라 손으로 계산한 값과 Python 결과를 비교한다.

```python
samples=[(120,120,120),(255,0,0),(200,100,50)]
for r,g,b in samples:
    brightness=(r+g+b)//3
    print((r,g,b),'→',brightness)
```

## Q08. 왜 `/`가 아니라 `//`를 사용하는가?

### 답
문제에서 밝기와 threshold를 소수점이 없는 정수로 계산하도록 정수 나눗셈을 사용하기 때문이다.

### 상세 설명
Python의 `/`는 일반 나눗셈이어서 결과가 `2.333...`처럼 실수가 될 수 있다. 반면 `//`는 몫만 남기고 소수점 아래 부분을 버린다. 예를 들어 `7/3`은 약 2.333이고 `7//3`은 2다. 이 문제에서 `(R+G+B)//3`과 `total//pixel_count`를 사용하는 이유도 결과를 정수로 만들기 위해서다.

`//`는 단순히 반올림이 아니라 **버림에 가까운 몫 계산**이라는 점도 중요하다. 예를 들어 305를 3으로 나누면 실제 평균은 약 101.66이지만 `305//3`은 101이다. 이 작은 차이가 threshold와 경계 픽셀의 분류 결과를 바꿀 수도 있다.

### 왜 필요한가?
문제에서 지정한 계산 방식과 정확히 같은 결과를 만들기 위해 필요하다.

### 헷갈리기 쉬운 점
`//`는 반올림이 아니다.

### 직접 해보기
`305/3`과 `305//3`을 비교하고 자료형도 출력한다.

```python
a=305/3
b=305//3
print(a, type(a))
print(b, type(b))
```

이 예시는 '색이 강하다'와 '밝기 값이 크다'를 같은 개념으로 보면 안 된다는 것을 보여준다. 문제에서 정의한 계산법만 놓고 보면 서로 전혀 다른 색들이 같은 밝기를 가질 수 있다.

### 왜 필요한가?
RGB와 brightness 사이의 관계를 실제 숫자로 이해하기 좋은 대표 예제다.

### 헷갈리기 쉬운 점

### 직접 해보기

```python
colors=[('red',(255,0,0)),('green',(0,255,0)),('blue',(0,0,255))]
for name,rgb in colors:
    print(name, sum(rgb)//3)
```

## Q10. 밝기 계산을 함수로 분리하는 이유는 무엇인가?

같은 계산을 반복해서 쓰지 않고, 역할을 분리해 읽기·테스트·수정하기 쉽게 만들기 위해서다.

### 상세 설명
메인 코드 안에서 매 픽셀마다 `(r+g+b)//3`을 직접 써도 프로그램은 동작할 수 있다. 하지만 `calc_brightness(r,g,b)`라는 함수를 만들면 '이 부분은 픽셀 하나의 밝기를 계산하는 곳'이라는 의미가 분명해진다. 코드를 읽는 사람은 함수 이름만 보고도 역할을 예상할 수 있다.

또한 함수는 독립적으로 테스트할 수 있다. 전체 입력 과정을 실행하지 않고도 `calc_brightness(255,0,0)`을 호출해서 85가 나오는지 확인할 수 있다. 만약 나중에 밝기 공식을 수정해야 한다면 메인 코드 곳곳을 찾을 필요 없이 함수 내부만 바꾸면 된다.

### 왜 필요한가?
기능별로 나누면 오류가 발생한 위치를 찾기 쉽고 설명서 작성도 쉬워진다.

### 헷갈리기 쉬운 점
함수를 만드는 목적은 코드 줄 수를 무조건 줄이는 것이 아니라 **책임을 분리하는 것**이다.

### 직접 해보기
함수에 여러 RGB 값을 넣어 예상값과 실제값을 비교한다.

```python
def calc_brightness(r,g,b):
    return (r+g+b)//3

tests=[((255,0,0),85),((0,0,0),0),((255,255,255),255)]
for rgb,expected in tests:
    actual=calc_brightness(*rgb)
    print(rgb,'expected=',expected,'actual=',actual)
```

## Q11. H와 W는 각각 무엇을 뜻하는가?

### 답
H는 이미지의 높이, 즉 행(row)의 개수이고 W는 이미지의 너비, 즉 한 행에 들어 있는 픽셀의 개수다.

### 상세 설명
이미지를 표처럼 생각하면 이해하기 쉽다. H는 세로로 몇 줄 있는지를 나타내고 W는 각 줄에 가로로 몇 칸 있는지를 나타낸다. 입력 첫 줄이 `3 3`이라면 3행 3열 이미지이고 총 9개의 픽셀이 존재한다. `2 4`라면 2행 4열, 총 8픽셀이다.

프로그램에서는 H를 이용해 RGB 입력을 몇 줄 받아야 하는지 결정하고, W와 함께 `H*W`를 계산해 전체 픽셀 수를 구한다. 그래서 H와 W는 단순한 크기 정보가 아니라 반복 횟수와 평균 계산 모두에 사용된다.

### 왜 필요한가?
입력 구조와 전체 픽셀 수를 정확히 결정하는 기본 정보다.

### 헷갈리기 쉬운 점
H와 W 순서를 뒤집으면 행과 열을 혼동할 수 있다.

### 직접 해보기
`H=2, W=4`인 이미지를 종이에 2줄 4칸으로 그려 본다.

```python
H,W=2,4
print('행 수:',H)
print('한 행 픽셀 수:',W)
print('전체 픽셀:',H*W)
```

## Q12. 전체 픽셀 수는 왜 `H * W`인가?

### 답
각 행마다 W개의 픽셀이 있고 그런 행이 H개 있기 때문이다.

### 상세 설명
예를 들어 3행 4열 이미지를 생각하면 첫 번째 행에 4개, 두 번째 행에 4개, 세 번째 행에 4개가 있으므로 `4+4+4 = 12`개다. 이것을 곱셈으로 표현하면 `3*4 = 12`가 된다. 그래서 일반적으로 전체 픽셀 수는 `H*W`다.

이 값은 threshold를 계산할 때 매우 중요하다. 평균은 '전체 합 ÷ 전체 개수'이므로 밝기 합만 알고 있어서는 부족하고 픽셀 개수도 정확히 알아야 한다. W를 잘못 읽거나 행의 픽셀 수가 실제 W와 다르면 평균이 잘못 계산될 수 있다.

### 왜 필요한가?
threshold 계산의 분모이며 입력 검증 기준이 된다.

### 헷갈리기 쉬운 점
`H+W`가 아니라 `H*W`다.

### 직접 해보기
여러 H, W 조합으로 픽셀 수를 계산한다.

```python
for H,W in [(1,1),(2,5),(3,3),(100,100)]:
    print(H,'x',W,'=',H*W,'pixels')
```

## Q17. `map(int, ...)`는 왜 사용하는가?

### 답
여러 문자열에 같은 `int()` 변환을 한 번에 적용하기 편하기 때문이다.

### 상세 설명
`pixel.split(',')` 결과가 `['255','0','0']`이라면 각각 `int(parts[0])`, `int(parts[1])`, `int(parts[2])`로 바꿀 수도 있다. 하지만 `map(int, parts)`를 사용하면 각 원소에 `int` 함수를 차례대로 적용한다. 이를 `r, g, b = map(int, pixel.split(','))`처럼 쓰면 분리와 정수 변환을 간결하게 한 줄로 표현할 수 있다.

다만 초보자에게는 한 줄이 너무 압축되어 보일 수 있으므로 처음에는 `split → map → 대입` 순서를 따로 이해한 뒤 한 줄로 합치는 편이 좋다.

### 왜 필요한가?
반복되는 형 변환 코드를 줄이고 입력 처리 코드를 간결하게 만든다.

### 헷갈리기 쉬운 점
`map()` 자체는 리스트가 아니며 필요하면 `list()`로 감싸 내용을 확인할 수 있다.

### 직접 해보기
같은 값을 개별 `int()` 방식과 `map()` 방식으로 변환해 비교한다.

```python
pixel='255,0,0'
parts=pixel.split(',')
manual=[int(parts[0]),int(parts[1]),int(parts[2])]
mapped=list(map(int,parts))
print(manual)
print(mapped)
```

입력 RGB는 한 번 읽고 나면 다시 자동으로 나타나지 않는다. 그런데 threshold는 모든 픽셀의 밝기를 먼저 알아야 계산할 수 있고, threshold가 계산된 뒤에는 각 픽셀의 밝기를 다시 비교해야 한다. 따라서 계산된 brightness들을 저장해 두는 방식이 이해하기 쉽다.

2차원 리스트를 사용하면 첫 행의 밝기와 둘째 행의 밝기가 분리되어 있으므로 최종 결과를 행 단위로 만들기도 편하다. 예제에서는 `[[255,200,0],[120,60,30],[85,85,85]]`처럼 저장된다.

### 왜 필요한가?
두 번째 단계인 threshold 비교에서 다시 사용할 중간 데이터를 보존한다.

### 헷갈리기 쉬운 점
평균만 구한다고 바로 값을 버리면 나중에 0/1 변환을 위해 원래 밝기가 다시 필요하다.

### 직접 해보기

## Q22. 전체 밝기 합을 구할 때 왜 `total = 0`으로 시작하는가?

### 답
아직 어떤 밝기도 더하지 않은 초기 합계가 0이기 때문이다.

### 상세 설명
누적 합계를 구할 때는 시작값이 필요하다. 숫자들의 합을 구하는 경우 가장 자연스러운 시작값은 0이다. 이후 각 밝기를 읽을 때마다 `total += brightness`를 실행하면 현재까지 본 모든 밝기의 합이 `total`에 저장된다.

예를 들어 값이 255, 200, 0이라면 처음 total은 0, 첫 번째 값을 더한 뒤 255, 두 번째 뒤 455, 세 번째 뒤에도 455가 된다. 이처럼 변수 하나가 반복될 때마다 갱신되는 것을 누적(accumulation)이라고 볼 수 있다.

### 왜 필요한가?
threshold는 전체 밝기의 평균이므로 먼저 전체 합이 필요하다.

### 헷갈리기 쉬운 점
`total`을 반복문 안에서 매번 0으로 초기화하면 이전까지 더한 값이 사라진다.

### 직접 해보기
값을 하나씩 더할 때 total이 어떻게 변하는지 출력한다.

```python
values=[255,200,0]
total=0
for v in values:
    total += v
    print('v=',v,'total=',total)
```

## Q23. 2차원 밝기 맵의 전체 합은 어떻게 계산하는가?

### 답
각 행을 반복하고 그 행 안의 각 밝기를 다시 반복하면서 `total`에 더한다.

### 상세 설명
`brightness_map`은 리스트 안에 리스트가 있는 2차원 구조이므로 모든 값을 더하려면 두 겹의 반복문을 사용할 수 있다. 바깥 반복문에서 `row`를 하나씩 꺼내고 안쪽 반복문에서 `brightness`를 하나씩 꺼낸 뒤 `total += brightness`를 실행한다.

Python의 `sum()`을 이용해 각 행의 합을 구하고 그것들을 다시 더하는 방식도 가능하다. 하지만 과제에서 중첩 반복문의 흐름을 이해하는 것이 중요하므로 처음에는 직접 모든 값을 순회하는 방식이 학습에 도움이 된다.

### 왜 필요한가?
전체 밝기 합이 있어야 평균 threshold를 계산할 수 있다.

### 헷갈리기 쉬운 점
행 하나의 합만 구하고 전체 합이라고 착각하지 않도록 한다.

### 직접 해보기
예제 밝기 맵을 중첩 반복문으로 모두 더해 920이 나오는지 확인한다.

```python
brightness_map=[[255,200,0],[120,60,30],[85,85,85]]
total=0
for row in brightness_map:
    for brightness in row:
        total += brightness
print('전체 합:',total)
```

## Q24. threshold는 어떻게 계산하며 예제에서는 왜 102인가?

### 답
모든 픽셀 밝기의 합을 전체 픽셀 수로 정수 나눗셈한 값이다. 예제에서는 `920 // 9 = 102`다.

### 상세 설명
threshold는 이 문제에서 밝은 픽셀과 어두운 픽셀을 나누는 기준선이다. 계산식은 `total // (H * W)`다. 예제의 밝기 맵 전체 합은 920이고 이미지 크기는 3×3이므로 픽셀 수는 9개다. 따라서 `920 // 9`를 계산해 102를 얻는다. 실제 나눗셈 값은 약 102.22지만 정수 나눗셈이므로 소수점 아래를 버린다.

threshold가 계산되기 전에는 어떤 픽셀이 1인지 0인지 최종적으로 결정할 수 없다. 예를 들어 밝기 100은 threshold가 90이면 1이지만 threshold가 110이면 0이 된다. 즉 픽셀의 절대 밝기보다 **전체 이미지와 비교했을 때 어느 쪽인지**가 중요하다.

### 왜 필요한가?
최종 0/1 분류에 사용되는 핵심 기준값이다.

### 헷갈리기 쉬운 점
밝기 한 개의 평균이 아니라 **모든 픽셀 밝기의 평균**이다.

### 직접 해보기
예제 전체 합과 픽셀 수를 직접 출력하고 threshold를 계산한다.

```python
total=920
H=W=3
pixel_count=H*W
threshold=total//pixel_count
print('합=',total,'픽셀 수=',pixel_count,'threshold=',threshold)
```

## Q25. 왜 threshold를 전체 평균 밝기로 정하는가?

### 답
이미지 전체의 일반적인 밝기를 기준선으로 삼아 상대적으로 밝은 픽셀과 어두운 픽셀을 나누기 위해서다.

### 상세 설명
평균은 전체 데이터가 대체로 어느 수준에 있는지 보여주는 가장 기본적인 대표값 중 하나다. 이 문제에서는 모든 픽셀의 brightness를 평균내어 '이 이미지에서 보통 밝기는 이 정도다'라는 기준을 만든다. 그 기준 이상인 픽셀은 1, 미만인 픽셀은 0으로 분류한다.

예를 들어 아주 어두운 사진은 threshold도 낮아질 수 있고, 전체적으로 밝은 사진은 threshold가 높아질 수 있다. 따라서 고정된 숫자 128을 무조건 사용하는 방식과 달리 이미지 데이터 자체에 따라 기준값이 바뀐다. 이것을 간단한 형태의 자동 threshold라고 생각할 수 있다.

### 왜 필요한가?
왜 모든 픽셀을 먼저 읽고 나서야 0/1 결과를 결정할 수 있는지 이해하게 해준다.

### 헷갈리기 쉬운 점
평균보다 '밝다'는 것은 절대적으로 밝다는 뜻이 아니라 **그 이미지 안에서 상대적으로 밝다**는 의미다.

### 직접 해보기
어두운 값들만 있는 경우와 밝은 값들만 있는 경우 평균이 어떻게 달라지는지 비교한다.

```python
for values in [[10,20,30],[200,220,240]]:
    threshold=sum(values)//len(values)
    print(values,'threshold=',threshold)
```

## Q75. 다시 처음부터 풀게 된다면 어떤 순서로 접근하는 것이 좋은가?

### 답
문제 목표 확인 → 입력 구조 분석 → 한 픽셀 밝기 계산 → 한 행 처리 → 전체 밝기 저장 → 평균 threshold 계산 → 0/1 변환 → 예제 추적 → 경계값 테스트 순서로 접근한다.

### 상세 설명
처음부터 최종 코드를 한 번에 작성하기보다 한 단계씩 작은 예제로 검증하면 오류 위치를 찾기 쉽다. 특히 입력 파싱과 threshold 경계 조건은 작은 실수 하나가 전체 결과를 바꾸므로 손으로 먼저 예제를 계산한 뒤 구현하는 방식이 안전하다.

### 마지막 정리
처음에는 RGB, 밝기, threshold가 서로 따로 보였지만 지금은 하나의 데이터 처리 흐름으로 연결된다. 다시 풀 때는 문법을 외우기보다 각 단계의 입력값과 출력값을 확인하면서 구현한다.

```python
order=['목표','입력','밝기','저장','threshold','0/1','출력','테스트']
print(' → '.join(order))
```
