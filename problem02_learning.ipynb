## 실행 방법
1. 위에서 아래로 `Run All`을 실행한다.

3. 예상과 실제가 다르면 최초로 달라진 중간 변수를 찾는다.
4. 모든 `assert`가 통과하고 마지막 셀까지 멈추지 않으면 전체 실험이 재현된 것이다.

## Q01. 문제 2가 근본적으로 원하는 것은 무엇인가?

> **학습 단계:** RGB와 밝기 계산  
> **이번 질문의 목표:** 한 픽셀의 RGB를 정수 밝기 하나로 바꾸는 원리를 작은 값부터 확인한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
컬러 이미지의 RGB 데이터를 각 픽셀의 밝기 값으로 바꾸고, 전체 평균 밝기를 기준으로 각 픽셀을 0 또는 1로 분류하는 것이다.

### 이전 단계와 연결
첫 질문이므로 먼저 문제 전체의 입력→처리→출력 흐름을 잡는다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
이 문제는 실제 JPG나 PNG 파일을 편집하는 문제가 아니라 숫자로 표현된 픽셀 데이터를 계산하는 알고리즘 문제다. 입력은 R,G,B 세 값으로 들어오고, 중간에는 각 픽셀의 밝기라는 하나의 정수로 바뀐다. 그다음 모든 밝기의 평균을 threshold로 정하고, 각 픽셀이 그 기준 이상인지 미만인지 판단한다. 마지막 결과는 원래 이미지와 같은 행·열 구조를 유지한 0과 1의 문자열이다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
여기서 흑백 변환은 실제 이미지 파일 편집이 아니라 숫자 데이터 분류다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
`RGB → 밝기 → threshold → 0/1` 순서를 적고 각 단계에서 만들어지는 값을 써 본다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **컬러 이미지의 RGB 데이터를 각 픽셀의 밝기 값으로 바꾸고, 전체 평균 밝기를 기준으로 각 픽셀을 0 또는 1로 분류하는 것이다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q02에서 다음 개념과 연결한다.


```python
print('=== Q01: 문제 2가 근본적으로 원하는 것은 무엇인가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
print('RGB → brightness → threshold → 0/1')
print('[확인] Q01 셀 실행 완료')
```

## Q02. RGB는 무엇이고 왜 세 값이 필요한가?

> **학습 단계:** RGB와 밝기 계산  
> **이번 질문의 목표:** 한 픽셀의 RGB를 정수 밝기 하나로 바꾸는 원리를 작은 값부터 확인한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
RGB는 Red, Green, Blue의 약자로 픽셀 하나의 색을 나타내는 세 개의 색 성분이다.

### 헷갈린 점과 확인 방법

## Q03. RGB 값은 왜 0부터 255까지인가?

> **학습 단계:** RGB와 밝기 계산  
> **이번 질문의 목표:** 한 픽셀의 RGB를 정수 밝기 하나로 바꾸는 원리를 작은 값부터 확인한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
일반적인 8비트 색상 성분은 0부터 255까지 총 256단계를 표현할 수 있기 때문이다.

### 이전 단계와 연결
Q02에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
8비트는 0과 1을 8자리 사용한다는 뜻이고 가능한 조합 수는 `2**8 = 256`개다. 정수 범위로 표현하면 0부터 255까지다. 0은 해당 색 성분이 없는 상태, 255는 가장 강한 상태로 생각하면 된다. 따라서 이 문제에서 계산되는 밝기도 기본적으로 0~255 범위를 벗어나지 않는다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
256단계이지만 최댓값은 256이 아니라 255다.

### 직접 실행해 볼 내용
`2**8`과 `2**8-1`을 실행한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **일반적인 8비트 색상 성분은 0부터 255까지 총 256단계를 표현할 수 있기 때문이다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

## Q04. 한 픽셀 문자열 `255,0,0`은 실제로 어떤 구조인가?

- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
서로 다른 색도 같은 밝기 값을 가질 수 있다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
빨강 `(255,0,0)`과 회색 `(85,85,85)`의 밝기를 비교한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **흑백 분류에서는 색의 종류보다 얼마나 밝고 어두운지를 하나의 값으로 비교하는 것이 편하기 때문이다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q07에서 다음 개념과 연결한다.


```python
print('=== Q06: 왜 RGB 세 값을 밝기 하나로 바꾸는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
for rgb in [(255,0,0),(85,85,85)]:
    print(rgb, sum(rgb)//3)
print('[확인] Q06 셀 실행 완료')
```

- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
원색 세 가지는 이 공식에서는 모두 같은 밝기 85가 된다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
빨강, 초록, 파랑의 밝기를 비교한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **`(255+0+0)//3 = 85`이기 때문이다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q10에서 다음 개념과 연결한다.


```python
print('=== Q09: 빨간색 `(255,0,0)`의 밝기는 왜 85인가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
for rgb in [(255,0,0),(0,255,0),(0,0,255)]:
    print(rgb, sum(rgb)//3)
print('[확인] Q09 셀 실행 완료')
```

## Q10. 밝기 계산을 함수로 분리하는 이유는 무엇인가?

> **이번 질문의 목표:** 한 픽셀의 RGB를 정수 밝기 하나로 바꾸는 원리를 작은 값부터 확인한다.

코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
H+W가 아니라 H×W다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
2행 5열과 10행 10열의 픽셀 수를 계산한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **H개의 행마다 W개의 픽셀이 있기 때문이다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q13에서 다음 개념과 연결한다.


```python
print('=== Q12: 전체 픽셀 수는 왜 `H * W`인가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
for H,W in [(2,5),(10,10)]:
    print(H,W,'->',H*W)
print('[확인] Q12 셀 실행 완료')
```

## Q13. 첫 줄 `3 3`은 어떻게 두 변수에 저장하는가?

> **학습 단계:** 입력 구조와 2차원 밝기 맵  
> **이번 질문의 목표:** 문자열 한 개, 한 행, 전체 이미지 순서로 범위를 넓혀 입력 구조를 확인한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
공백 기준으로 나눈 뒤 정수로 바꿔 `height, width`에 저장한다.

`input().split()`은 입력 문자열을 공백 기준으로 나눠 두 문자열을 만든다. `map(int, ...)`는 그 문자열들에 `int()`를 적용해 정수로 바꾼다. 마지막으로 언패킹을 사용해 첫 값을 height, 두 번째 값을 width에 저장한다.

- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
`split()`까지만 하면 여전히 문자열이다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용

- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
첫 split은 픽셀 단위, 두 번째 split은 색 성분 단위다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
문자열을 나눈 뒤 세 조각인지 확인한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **`split(',')`으로 쉼표 기준 세 조각으로 나눈다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q16에서 다음 개념과 연결한다.


```python
print('=== Q15: RGB 문자열 `255,0,0`은 어떻게 세 값으로 나누는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
pixel='255,0,0'
parts=pixel.split(',')
print(parts,len(parts))
print('[확인] Q15 셀 실행 완료')
```

## Q16. 문자열 숫자는 왜 `int()`로 바꾸는가?

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

## Q17. `map(int, ...)`는 왜 사용하는가?

- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
바깥 반복문과 안쪽 반복문의 역할이 다르다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
2×2 숫자 리스트를 중첩 반복문으로 출력한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **이미지가 행과 열의 2차원 구조라 모든 행과 모든 픽셀을 차례로 확인해야 하기 때문이다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q19에서 다음 개념과 연결한다.


```python
print('=== Q18: 한 행에서 전체 이미지로 확장하기 ===')
rgb_rows = [
    ['255,255,255', '200,200,200', '0,0,0'],
    ['120,120,120', '60,60,60', '30,30,30'],
]
brightness_map = []

for row_index, rgb_row in enumerate(rgb_rows):
    brightness_row = []
    print(f'[{row_index}행 시작] 원본 =', rgb_row)
    for column_index, pixel_text in enumerate(rgb_row):
        r, g, b = map(int, pixel_text.split(','))
        brightness = (r + g + b) // 3
        brightness_row.append(brightness)
        print(f'  ({row_index}, {column_index}) RGB={(r,g,b)} -> 밝기={brightness}')
    brightness_map.append(brightness_row)
    print(f'[{row_index}행 완료] 밝기 행 =', brightness_row)

print('전체 밝기 맵 =', brightness_map)
assert brightness_map == [[255, 200, 0], [120, 60, 30]]
```

## Q19. `brightness_row`를 매 행마다 새로 만드는 이유는 무엇인가?

- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
`brightness_map[1][0]`은 두 번째 행 첫 번째 값이다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
2차원 리스트에서 특정 위치 값을 꺼내 본다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

## Q24. threshold는 정확히 무엇인가?

> **학습 단계:** 평균 임계값과 흑백 변환  
> **이번 질문의 목표:** 누적 합과 정수 평균을 구한 뒤 각 밝기를 0 또는 1로 분류한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
이미지 전체 픽셀 밝기의 정수 평균이며, 밝고 어두움을 나누는 기준값이다.

### 이전 단계와 연결
Q23에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
모든 픽셀의 밝기를 더한 뒤 전체 픽셀 수로 나누면 이미지의 평균적인 밝기를 얻을 수 있다. 이 문제는 그 평균값을 threshold라고 부르고, 각 픽셀을 0 또는 1로 분류하는 기준으로 사용한다. threshold 자체도 최종 출력의 첫 줄에 출력해야 한다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
threshold는 미리 정해진 128 같은 고정값이 아니라 입력 이미지마다 달라진다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
예제 밝기들의 평균을 계산한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리

### 이전 단계와 연결

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.

### 헷갈린 점과 확인 방법
920/9는 약 102.22지만 문제에서는 `//`를 써서 102다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

9개 값을 직접 합산하고 9로 나눈다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q26에서 다음 개념과 연결한다.


```python
print('=== Q25: 예제 임계값 102를 단계별로 확인 ===')
brightness_map = [
    [255, 200, 0],
    [120, 60, 30],
    [85, 85, 85],
]
total = 0
count = 0

for row_index, brightness_row in enumerate(brightness_map):
    row_sum = sum(brightness_row)
    total += row_sum
    count += len(brightness_row)
    print(f'{row_index}행 합={row_sum}, 현재 누적 합={total}, 현재 개수={count}')

threshold = total // count
print('전체 밝기 합 =', total)
print('전체 픽셀 수 =', count)
print(f'threshold = {total} // {count} = {threshold}')
assert total == 920
assert count == 9
assert threshold == 102
```

전체 픽셀 수를 `height * width`로 계산하기 위해서다.

### 이전 단계와 연결
Q26에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
밝기 맵만 보고도 픽셀 수를 구할 수 있지만 현재 코드에서는 문제 입력으로 받은 height와 width를 함수에 명시적으로 전달한다. 이렇게 하면 '전체 픽셀 수는 이미지 높이×너비'라는 문제 구조가 코드에 그대로 드러난다. 동시에 함수 안에서 평균 계산이 어떤 기준으로 이루어지는지도 이해하기 쉽다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
height와 width는 밝기 값이 아니라 이미지 크기 정보다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
2행 3열에서 height*width와 실제 원소 수를 비교한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **전체 픽셀 수를 `height * width`로 계산하기 위해서다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q28에서 다음 개념과 연결한다.


```python
print('=== Q27: threshold 계산 함수에서는 왜 `height`와 `width`를 받는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
m=[[10,20,30],[40,50,60]]
height,width=2,3
print(height*width)
print(sum(len(row) for row in m))
print('[확인] Q27 셀 실행 완료')
```

## Q30. 흑백 결과를 문자열로 만드는 이유는 무엇인가?

> **학습 단계:** 평균 임계값과 흑백 변환  
> **이번 질문의 목표:** 누적 합과 정수 평균을 구한 뒤 각 밝기를 0 또는 1로 분류한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
한 행의 0과 1을 공백 없이 이어 붙여 출력하기 가장 편하기 때문이다.

### 이전 단계와 연결
Q29에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
최종 출력은 `[1,1,0]`처럼 리스트 모양이 아니라 `110`처럼 붙어 있는 문자열이어야 한다. 그래서 각 행마다 빈 문자열을 만들고 조건에 따라 `'1'` 또는 `'0'`을 덧붙이면 출력 형식에 바로 맞출 수 있다. 숫자 리스트를 만든 뒤 나중에 문자열로 변환하는 방법도 있지만 단계가 하나 더 필요하다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
문자열 `'1'`과 정수 `1`은 다르다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
리스트와 문자열의 출력 차이를 비교한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **한 행의 0과 1을 공백 없이 이어 붙여 출력하기 가장 편하기 때문이다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q31에서 다음 개념과 연결한다.


```python
print('=== Q30: 흑백 결과를 문자열로 만드는 이유는 무엇인가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
print([1,1,0])
print('110')
print('[확인] Q30 셀 실행 완료')
```

이번 질문의 핵심은 **문제에서 threshold 이상을 밝은 쪽으로 분류하고 그 값을 1로 표현하도록 정했기 때문이다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q33에서 다음 개념과 연결한다.


```python
print('=== Q32: 밝기가 threshold 이상이면 왜 `1`을 넣는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
threshold=102
for b in [120,85,102]:
    print(b,'->',1 if b>=threshold else 0)
print('[확인] Q32 셀 실행 완료')
```

## Q33. 왜 `>`가 아니라 `>=`를 사용하는가?

> **학습 단계:** 평균 임계값과 흑백 변환  
> **이번 질문의 목표:** 누적 합과 정수 평균을 구한 뒤 각 밝기를 0 또는 1로 분류한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
threshold와 정확히 같은 밝기도 1로 처리해야 하기 때문이다.

### 이전 단계와 연결

### 상세 설명
문제 표현이 'threshold 이상'이므로 같은 값도 포함해야 한다. `>`를 쓰면 threshold와 같은 픽셀은 0이 되어 문제 조건과 달라진다. 특히 전체가 같은 밝기인 이미지에서는 모든 픽셀이 threshold와 같아지므로 이 차이가 결과 전체를 바꾼다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

작은 밝기 맵과 threshold를 넣어 본다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **`brightness_map`과 `threshold`를 입력받아 0/1 문자열 행들의 리스트를 반환한다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q36에서 다음 개념과 연결한다.


```python
print('=== Q35: 밝기 맵을 흑백 문자열로 변환 ===')
def to_black_white(brightness_map, threshold):
    result = []
    for row_index, brightness_row in enumerate(brightness_map):
        result_row = ''
        print(f'{row_index}행 변환 시작:', brightness_row)
        for column_index, brightness in enumerate(brightness_row):
            bit = '1' if brightness >= threshold else '0'
            result_row += bit
            relation = '>=' if brightness >= threshold else '<'
            print(f'  ({row_index}, {column_index}) {brightness} {relation} {threshold} -> {bit}')
        result.append(result_row)
        print(f'{row_index}행 변환 완료:', result_row)
    return result

sample_map = [[255, 200, 0], [120, 60, 30], [85, 85, 85]]
sample_threshold = 102
sample_result = to_black_white(sample_map, sample_threshold)
print('최종 결과 =', sample_result)
assert sample_result == ['110', '100', '000']
```

## Q36. `main()` 함수는 왜 필요한가?

> **학습 단계:** 함수화와 전체 흐름 연결  
> **이번 질문의 목표:** 앞에서 따로 실험한 계산을 함수로 묶고 예제 입력 전체에 적용한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
입력부터 출력까지 프로그램의 전체 실행 순서를 한곳에 모으기 위해서다.

### 이전 단계와 연결
Q35에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
밝기 계산, threshold 계산, 흑백 변환을 각각 함수로 나눴다면 누군가는 이 함수들을 올바른 순서로 호출해야 한다. `main()`은 입력을 받고 필요한 함수를 호출한 뒤 최종 출력을 담당하는 중심 역할을 한다. 이렇게 하면 프로그램의 큰 흐름과 세부 기능이 분리되어 읽기 쉬워진다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
main은 특별한 예약어가 아니라 우리가 정한 함수 이름이다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
작은 main 함수에서 두 함수를 순서대로 호출한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **입력부터 출력까지 프로그램의 전체 실행 순서를 한곳에 모으기 위해서다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

## Q37. `if __name__ == '__main__':`는 왜 사용하는가?

색이 달라도 밝기는 같을 수 있다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
예제 문자열을 직접 파싱해 밝기 맵을 만든다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **`[[255,200,0],[120,60,30],[85,85,85]]`가 된다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q39에서 다음 개념과 연결한다.


```python
print('=== Q38: RGB 문자열에서 밝기 맵까지 누적 실행 ===')
height, width = 3, 3
row_texts = [
    '255,255,255 200,200,200 0,0,0',
    '120,120,120 60,60,60 30,30,30',
    '255,0,0 0,255,0 0,0,255',
]
brightness_map = []

for row_index, row_text in enumerate(row_texts):
    pixel_texts = row_text.split()
    print(f'{row_index}행 픽셀 문자열 =', pixel_texts)
    assert len(pixel_texts) == width
    brightness_row = []
    for column_index, pixel_text in enumerate(pixel_texts):
        rgb = tuple(map(int, pixel_text.split(',')))
        brightness = sum(rgb) // 3
        brightness_row.append(brightness)
        print(f'  ({row_index}, {column_index}) {pixel_text} -> {rgb} -> {brightness}')
    brightness_map.append(brightness_row)

print('완성된 밝기 맵 =', brightness_map)
assert len(brightness_map) == height
assert brightness_map == [[255, 200, 0], [120, 60, 30], [85, 85, 85]]
```

- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
255보다 커야 1이 되는 것이 아니라 255와 같아도 1이다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
2×2 흰색 이미지를 계산한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **모든 밝기가 255이고 threshold도 255이므로 모든 픽셀이 threshold 이상이기 때문이다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q42에서 다음 개념과 연결한다.


```python
print('=== Q41: 전체가 흰색인 이미지에서는 왜 결과가 모두 1인가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
m=[[255,255],[255,255]]
t=sum(sum(r) for r in m)//4
print(t)
print([''.join('1' if b>=t else '0' for b in r) for r in m])
print('[확인] Q41 셀 실행 완료')
```

### 다음 질문으로 연결

```python
print('=== Q43: 입력과 동시에 밝기와 합을 누적 ===')
row_texts = [
    '255,255,255 200,200,200 0,0,0',
    '120,120,120 60,60,60 30,30,30',
    '255,0,0 0,255,0 0,0,255',
]
brightness_map = []
total = 0
pixel_count = 0

for row_index, row_text in enumerate(row_texts):
    brightness_row = []
    for pixel_text in row_text.split():
        r, g, b = map(int, pixel_text.split(','))
        brightness = (r + g + b) // 3
        brightness_row.append(brightness)
        total += brightness
        pixel_count += 1
    brightness_map.append(brightness_row)
    print(f'{row_index}행 후: map={brightness_map}, total={total}, count={pixel_count}')

threshold = total // pixel_count
print('threshold =', threshold)
assert (total, pixel_count, threshold) == (920, 9, 102)
```

## Q44. `sum()`을 쓰면 코드를 더 짧게 만들 수 있는가?

> **학습 단계:** 함수화와 전체 흐름 연결  
> **이번 질문의 목표:** 앞에서 따로 실험한 계산을 함수로 묶고 예제 입력 전체에 적용한다.

### 먼저 떠올린 질문

### 답
가능하지만 중첩 리스트에서는 행별 합을 다시 합치는 방식이 필요하다.

### 이전 단계와 연결
Q43에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
`sum(sum(row) for row in brightness_map)`처럼 쓰면 중첩 반복문 없이 전체 합을 한 줄에 계산할 수 있다. 하지만 처음 배우는 단계에서는 중첩 반복문이 데이터 구조와 누적 과정을 더 명확하게 보여준다. 짧은 코드가 항상 더 이해하기 쉬운 것은 아니다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
`sum(brightness_map)`은 내부가 리스트라 그대로는 원하는 숫자 합이 되지 않는다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
반복문 방식과 sum 방식을 비교한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **가능하지만 중첩 리스트에서는 행별 합을 다시 합치는 방식이 필요하다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q45에서 다음 개념과 연결한다.


```python
print('=== Q44: `sum()`을 쓰면 코드를 더 짧게 만들 수 있는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
m=[[1,2,3],[4,5,6]]
total1=0
for row in m:
    for x in row:
        total1+=x
total2=sum(sum(row) for row in m)
print(total1,total2)
print('[확인] Q44 셀 실행 완료')
```

## Q45. 문자열 덧붙이기 대신 `join()`을 사용할 수도 있는가?

> **학습 단계:** 함수화와 전체 흐름 연결  
> **이번 질문의 목표:** 앞에서 따로 실험한 계산을 함수로 묶고 예제 입력 전체에 적용한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
가능하다. 각 픽셀 결과를 모은 뒤 `''.join(...)`으로 한 행 문자열을 만들 수 있다.

### 이전 단계와 연결

### 상세 설명
현재 코드는 `result_row += '1'`처럼 문자열을 하나씩 이어 붙인다. 다른 방식은 `'1'`과 `'0'`을 리스트에 모은 뒤 마지막에 join으로 합치는 것이다. 데이터가 매우 많다면 join 방식이 더 효율적일 수 있지만, 이 문제의 최대 크기에서는 둘 다 충분히 동작한다. 중요한 것은 두 방식 모두 최종 결과가 똑같아야 한다는 점이다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 헷갈린 점과 확인 방법
`join()`은 문자열 원소를 합치는 기능이므로 정수 리스트라면 먼저 문자열로 바꿔야 한다.

```python
print('=== Q45: 문자열 덧붙이기 대신 `join()`을 사용할 수도 있는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
row=[120,60,102]
t=102
a=''
for b in row:
    a += '1' if b>=t else '0'
b=''.join('1' if x>=t else '0' for x in row)
print(a,b,a==b)
print('[확인] Q45 셀 실행 완료')
```

> **이번 질문의 목표:** 앞에서 따로 실험한 계산을 함수로 묶고 예제 입력 전체에 적용한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
코드만 봐서는 이유가 바로 드러나지 않는 핵심 계산이나 조건 옆에 쓰는 것이 좋다.

### 이전 단계와 연결
Q46에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
모든 줄을 그대로 한국어로 번역하듯 주석 처리하면 오히려 코드가 지저분해질 수 있다. 대신 `// 3`을 쓰는 이유, threshold와 같은 값도 1이 되는 이유, 함수의 입력·출력처럼 중요한 의도를 설명하는 주석이 유용하다. 좋은 주석은 단순히 무엇을 하는가를 반복하기보다 왜 이렇게 하는가를 알려준다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
코드만 읽어도 명백한 `x += 1`에 'x에 1을 더한다' 같은 주석은 가치가 낮다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
밝기 공식 옆에 이유를 설명하는 주석을 붙인다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **코드만 봐서는 이유가 바로 드러나지 않는 핵심 계산이나 조건 옆에 쓰는 것이 좋다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q48에서 다음 개념과 연결한다.


```python
print('=== Q47: 주석은 어디에 쓰는 것이 좋은가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
r,g,b=255,0,0
# 과제에서 지정한 정수 평균 공식으로 픽셀 밝기를 계산한다.
brightness=(r+g+b)//3
print(brightness)
print('[확인] Q47 셀 실행 완료')
```

### 학습 정리
이번 질문의 핵심은 **예제 하나만 맞아도 다른 입력에서 오류가 날 수 있기 때문이다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q50에서 다음 개념과 연결한다.


```python
print('=== Q49: 서로 다른 성격의 테스트 케이스 ===')
def convert_brightness_map(brightness_map):
    total = sum(sum(row) for row in brightness_map)
    count = sum(len(row) for row in brightness_map)
    threshold = total // count
    result = [
        ''.join('1' if value >= threshold else '0' for value in row)
        for row in brightness_map
    ]
    return threshold, result

test_cases = [
    ('1x1 검정', [[0]], 0, ['1']),
    ('1x1 흰색', [[255]], 255, ['1']),
    ('균일 밝기', [[100, 100]], 100, ['11']),
    ('경계 포함', [[99, 100, 101]], 100, ['011']),
    ('최솟값/최댓값', [[0, 255]], 127, ['01']),
]

for name, brightness_map, expected_threshold, expected_result in test_cases:
    actual_threshold, actual_result = convert_brightness_map(brightness_map)
    print(f'{name}: map={brightness_map}')
    print(f'  예상=({expected_threshold}, {expected_result})')
    print(f'  실제=({actual_threshold}, {actual_result})')
    assert actual_threshold == expected_threshold
    assert actual_result == expected_result

print('[확인] 모든 테스트를 통과했다.')
```

## Q50. 예상 결과와 실제 결과를 왜 따로 적어야 하는가?

> **이번 질문의 목표:** 정상 예제뿐 아니라 1×1, 균일 영상, 임계값 경계와 잘못된 입력도 점검한다.

### 먼저 떠올린 질문

### 답
코드를 실행하기 전에 내가 예상한 논리가 맞는지 검증하기 위해서다.

### 이전 단계와 연결

먼저 예상 결과를 적고 실행 결과와 비교하면 코드가 우연히 맞아 보이는 것을 방지할 수 있다. 예상과 실제가 다르면 문제 이해가 잘못된 것인지 코드 구현이 잘못된 것인지 구분해서 원인을 찾을 수 있다. 테스트 기록에 PASS/FAIL을 적는 이유도 같은 맥락이다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상

- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.

### 헷갈린 점과 확인 방법

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **코드를 실행하기 전에 내가 예상한 논리가 맞는지 검증하기 위해서다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

## Q51. 이 문제의 시간 복잡도는 어떻게 생각할 수 있는가?

> **이번 질문의 목표:** 정상 예제뿐 아니라 1×1, 균일 영상, 임계값 경계와 잘못된 입력도 점검한다.

- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
변수 몇 개는 O(1)이지만 픽셀 수만큼 늘어나는 리스트가 O(H×W)다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
3×3과 100×100의 픽셀 수를 비교한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **brightness_map과 결과를 저장하므로 대략 O(H×W)의 추가 공간을 사용한다고 볼 수 있다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q53에서 다음 개념과 연결한다.


```python
print('=== Q52: 이 문제의 공간 복잡도는 어떻게 생각할 수 있는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
for H,W in [(3,3),(100,100)]:
    print(H*W)
print('[확인] Q52 셀 실행 완료')
```

## Q53. H와 W가 최대 100이면 픽셀은 최대 몇 개인가?

최대 `100 * 100 = 10,000`개다.

## Q55. RGB 성분 개수가 3개인지도 확인할 수 있는가?

> **학습 단계:** 경계값·예외·효율 검증  
> **이번 질문의 목표:** 정상 예제뿐 아니라 1×1, 균일 영상, 임계값 경계와 잘못된 입력도 점검한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
가능하다. `pixel.split(',')` 결과 길이가 3인지 검사하면 된다.

### 이전 단계와 연결
Q54에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
정상 픽셀은 R,G,B 세 값으로 구성된다. 입력이 `255,0`처럼 잘못되면 `r,g,b = ...`에서 언패킹할 값이 부족하다는 오류가 발생한다. 반대로 `1,2,3,4`처럼 값이 너무 많아도 같은 문제가 생긴다. 디버깅용으로 먼저 길이를 확인하면 어떤 픽셀에서 형식이 잘못됐는지 더 쉽게 찾을 수 있다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
정수 변환 전에 구조부터 확인하면 원인 파악이 쉽다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
정상/비정상 픽셀 문자열의 조각 수를 출력한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **가능하다. `pixel.split(',')` 결과 길이가 3인지 검사하면 된다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q56에서 다음 개념과 연결한다.


```python
print('=== Q55: RGB 성분 개수가 3개인지도 확인할 수 있는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
for pixel in ['255,0,0','255,0','1,2,3,4']:
    parts=pixel.split(',')
    print(pixel,parts,len(parts))
print('[확인] Q55 셀 실행 완료')
```

## Q56. RGB 값이 범위를 벗어나는지 검사하려면 어떻게 할 수 있는가?

> **학습 단계:** 경계값·예외·효율 검증  
> **이번 질문의 목표:** 정상 예제뿐 아니라 1×1, 균일 영상, 임계값 경계와 잘못된 입력도 점검한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
세 값 모두 `0 <= value <= 255`인지 확인하면 된다.

### 이전 단계와 연결
Q55에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명

3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수

- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
정수라고 해서 반드시 올바른 RGB 값인 것은 아니다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

정상 RGB와 범위를 벗어난 RGB를 검사한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q57에서 다음 개념과 연결한다.


```python
print('=== Q56: RGB 값이 범위를 벗어나는지 검사하려면 어떻게 할 수 있는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
for rgb in [(255,0,0),(300,0,0),(-1,10,20)]:
    valid=all(0<=v<=255 for v in rgb)
    print(rgb,valid)
print('[확인] Q56 셀 실행 완료')
```

## Q58. 평균 대신 중앙값을 threshold로 사용할 수도 있는가?

> **이번 질문의 목표:** 정상 예제뿐 아니라 1×1, 균일 영상, 임계값 경계와 잘못된 입력도 점검한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
다른 알고리즘이라면 가능하지만 이 과제에서는 평균을 사용해야 한다.

### 이전 단계와 연결
Q57에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
중앙값은 값을 정렬했을 때 가운데 값이라 극단값의 영향을 덜 받는 특징이 있다. 어떤 이미지 처리 문제에서는 중앙값이 더 적합할 수도 있지만, 현재 과제는 전체 밝기의 정수 평균을 threshold로 정의했다. 따라서 다른 기준을 쓰면 문제 조건을 어기게 된다. 같은 데이터에서 평균과 중앙값이 얼마나 달라질 수 있는지 비교해 보면 threshold 선택이 결과에 큰 영향을 줄 수 있다는 점도 이해할 수 있다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
더 좋은 통계 기준을 찾는 문제가 아니라 주어진 기준을 구현하는 문제다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
극단값이 있는 리스트에서 평균과 중앙값을 비교한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **다른 알고리즘이라면 가능하지만 이 과제에서는 평균을 사용해야 한다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q59에서 다음 개념과 연결한다.


```python
print('=== Q58: 평균 대신 중앙값을 threshold로 사용할 수도 있는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
values=[0,0,0,0,255]
avg=sum(values)//len(values)
median=sorted(values)[len(values)//2]
print('평균:',avg,'중앙값:',median)
print('[확인] Q58 셀 실행 완료')
```

## Q59. threshold가 0이 될 수 있는가?

> **학습 단계:** 경계값·예외·효율 검증  
> **이번 질문의 목표:** 정상 예제뿐 아니라 1×1, 균일 영상, 임계값 경계와 잘못된 입력도 점검한다.

### 답

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.

- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 다음 질문으로 연결

```python
print('=== Q59: threshold가 0이 될 수 있는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
values=[0,0,0,0]
t=sum(values)//len(values)
print(t,[1 if v>=t else 0 for v in values])
print('[확인] Q59 셀 실행 완료')
```

```python
print('=== Q60: threshold가 255가 될 수 있는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
values=[255,255,255,255]
t=sum(values)//len(values)
print(t,[1 if v>=t else 0 for v in values])
print('[확인] Q60 셀 실행 완료')
```

## Q61. 밝기 값이 threshold와 정확히 같은 픽셀은 어떤 의미인가?

> **학습 단계:** 경계값·예외·효율 검증  
> **이번 질문의 목표:** 정상 예제뿐 아니라 1×1, 균일 영상, 임계값 경계와 잘못된 입력도 점검한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
이미지 전체 평균과 정확히 같은 밝기를 가진 픽셀이고, 이 문제에서는 1로 분류된다.

### 이전 단계와 연결
Q60에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
평균과 같은 값은 밝은 쪽도 어두운 쪽도 아닌 경계값처럼 볼 수 있다. 하지만 문제는 경계값을 1 쪽에 포함하도록 `>=` 조건을 사용한다. 따라서 분류 규칙을 설명할 때 '평균보다 큰 픽셀'이라고만 말하면 정확하지 않고 '평균 이상'이라고 해야 한다. 자연어 설명 한 단어가 실제 코드 연산자와 연결된다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
'이상'과 '초과'를 같은 뜻으로 쓰면 안 된다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
99,100,101을 threshold 100과 비교한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **이미지 전체 평균과 정확히 같은 밝기를 가진 픽셀이고, 이 문제에서는 1로 분류된다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q62에서 다음 개념과 연결한다.


```python
print('=== Q61: 밝기 값이 threshold와 정확히 같은 픽셀은 어떤 의미인가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
t=100
for b in [99,100,101]:
    print(b,'1' if b>=t else '0')
print('[확인] Q61 셀 실행 완료')
```

오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **입력의 각 행마다 결과 행 하나를 만들기 때문이다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q64에서 다음 개념과 연결한다.


```python
print('=== Q63: 결과 행의 개수는 왜 항상 H와 같아야 하는가? ===')
# 실행 전에는 아래 print 결과를 먼저 예상하고, 실행 후 실제 값과 비교한다.
m=[[1,2],[3,4],[5,6]]
t=3
result=[''.join('1' if x>=t else '0' for x in row) for row in m]
print(result,len(result))
print('[확인] Q63 셀 실행 완료')
```

## Q64. 출력할 때 각 문자 사이에 공백을 넣으면 왜 안 되는가?

> **이번 질문의 목표:** 정상 예제뿐 아니라 1×1, 균일 영상, 임계값 경계와 잘못된 입력도 점검한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q66에서 다음 개념과 연결한다.


```python
print('=== Q65: input() 대신 같은 형식의 문자열로 재현 ===')
sample_input = '''3 3
255,255,255 200,200,200 0,0,0
120,120,120 60,60,60 30,30,30
255,0,0 0,255,0 0,0,255'''
input_lines = sample_input.strip().splitlines()
height, width = map(int, input_lines[0].split())
print('H, W =', height, width)

brightness_map = []
for row_index, line in enumerate(input_lines[1:]):
    pixel_texts = line.split()
    assert len(pixel_texts) == width
    brightness_row = []
    for pixel_text in pixel_texts:
        r, g, b = map(int, pixel_text.split(','))
        brightness_row.append((r + g + b) // 3)
    brightness_map.append(brightness_row)
    print(f'{row_index}행 밝기 =', brightness_row)

assert len(brightness_map) == height
total = sum(sum(row) for row in brightness_map)
threshold = total // (height * width)
result = [''.join('1' if value >= threshold else '0' for value in row) for row in brightness_map]
print('threshold =', threshold)
print('result =', result)
```

2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.

- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
셀 실행 순서와 화면에 보이는 셀 순서가 항상 같다고 가정하면 안 된다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **중간 셀부터 실행해도 이전 셀의 변수나 함수에 의존하지 않아 오류가 덜 난다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

## Q67. VS Code에서 ipynb 코드 셀이 실행되지 않을 때 무엇부터 확인해야 하는가?

> **이번 질문의 목표:** 학습용 실험과 제출용 코드를 구분하고 스스로 다시 구현할 수 있는지 확인한다.

`python problem02.py`라는 명령은 현재 작업 폴더에서 `problem02.py`를 찾는다. 터미널이 저장소 바깥 폴더나 한 단계 위 폴더에 있으면 파일이 실제로 존재해도 `No such file or directory` 오류가 날 수 있다. VS Code 탐색기에서 보이는 루트와 터미널 프롬프트의 경로를 비교하고, `dir`이나 `Get-Location`으로 현재 위치를 확인하면 원인을 빠르게 찾을 수 있다. 이 오류는 코드 내용과는 무관하다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
VS Code에서 파일이 열려 있다는 사실과 터미널 현재 폴더가 같다는 보장은 없다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
Python에서 현재 작업 폴더를 출력한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **터미널의 현재 작업 폴더와 실제 파일이 있는 폴더가 다를 수 있기 때문이다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

## Q72. AI 답변이 맞는지 어떻게 직접 검증할 수 있는가?

> **학습 단계:** 실행 환경·검증·최종 재구성  
> **이번 질문의 목표:** 학습용 실험과 제출용 코드를 구분하고 스스로 다시 구현할 수 있는지 확인한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
작은 예제와 경계값을 직접 실행해 예상 결과와 비교하면 된다.

### 이전 단계와 연결
Q71에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
AI가 설명한 공식이나 조건문이 그럴듯해 보여도 실제 문제 조건과 다를 수 있다. 그래서 손계산이 가능한 작은 입력을 만들고 코드 결과와 비교하는 것이 중요하다. 과제 예제, 1×1, 전체 검정, 전체 흰색, threshold와 정확히 같은 값이 포함된 경우는 특히 검증에 유용하다. 결과가 예상과 다르면 AI 답변, 내 이해, 코드 구현 세 가지를 다시 비교한다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
AI가 자신 있게 말해도 문제 원문과 실행 결과가 최종 기준이다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
밝기 함수의 기대값을 assert로 확인한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **작은 예제와 경계값을 직접 실행해 예상 결과와 비교하면 된다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
이 결과를 기억한 뒤 Q73에서 다음 개념과 연결한다.


```python
print('=== Q72: AI 설명도 반례로 직접 검증 ===')
def calc_brightness(r, g, b):
    return (r + g + b) // 3

checks = [
    ((0, 0, 0), 0),
    ((255, 255, 255), 255),
    ((255, 0, 0), 85),
    ((1, 1, 2), 1),
]

for rgb, expected in checks:
    actual = calc_brightness(*rgb)
    print(f'RGB={rgb}: 예상={expected}, 실제={actual}')
    assert actual == expected

print('[확인] 일반값뿐 아니라 최솟값, 최댓값, 나머지가 생기는 값도 통과했다.')
```

## Q73. `assert`는 테스트에서 어떻게 사용할 수 있는가?

> **학습 단계:** 실행 환경·검증·최종 재구성  
> **이번 질문의 목표:** 학습용 실험과 제출용 코드를 구분하고 스스로 다시 구현할 수 있는지 확인한다.

### 답
실제 결과가 기대값과 같은지 자동으로 확인하는 데 사용할 수 있다.

`assert 조건`은 조건이 참이면 조용히 넘어가고 거짓이면 `AssertionError`를 발생시킨다. 그래서 작은 함수의 결과를 빠르게 검증할 때 편리하다. 여러 개의 assert를 연속해서 두면 어느 테스트에서 조건이 깨졌는지 빠르게 찾을 수 있다. 다만 과제 제출 프로그램의 사용자 출력용으로 쓰기보다는 개발과 테스트 단계에서 사용하는 것이 적절하다.

## Q75. 75개 질문을 모두 학습한 뒤 문제를 다시 풀면 어떤 순서로 접근하는 것이 좋은가?

> **학습 단계:** 실행 환경·검증·최종 재구성  
> **이번 질문의 목표:** 학습용 실험과 제출용 코드를 구분하고 스스로 다시 구현할 수 있는지 확인한다.

### 먼저 떠올린 질문
코드를 바로 길게 작성하기 전에, 이번 질문에서 정말 확인하려는 대상이 값 하나인지, 한 행인지, 전체 이미지인지 구분한다. 또한 입력 자료형과 출력 자료형을 예상해 두면 실행 결과가 달라졌을 때 원인을 더 빨리 찾을 수 있다.

### 답
입력 구조 확인 → 한 픽셀 밝기 계산 → 밝기 맵 구성 → 전체 평균 threshold 계산 → 0/1 변환 → 출력 형식 확인 → 테스트 순서로 접근하면 된다.

### 이전 단계와 연결
Q74에서 확인한 내용을 출발점으로 삼아, 이번에는 범위를 한 단계 더 넓히거나 표현 방법을 바꿔 본다. 앞의 코드를 그대로 외우기보다 이번 셀에서 필요한 부분만 다시 작성하고, 새 변수와 출력문을 추가해 값이 만들어지는 순서를 추적한다.

### 상세 설명
처음부터 전체 코드를 한 번에 작성하려고 하면 입력 파싱, 평균 계산, 조건문, 출력 중 어디에서 틀렸는지 찾기 어렵다. 반대로 각 단계를 작은 함수나 작은 예제로 확인하면서 연결하면 오류가 생겨도 범위를 좁힐 수 있다. 먼저 RGB 한 픽셀이 제대로 숫자로 바뀌는지 확인하고, 그다음 한 행, 전체 이미지 순서로 범위를 넓힌다. 마지막에는 문제 예제뿐 아니라 1×1, 균일 이미지, threshold와 같은 값 같은 경계 테스트까지 실행해 기대값과 실제값이 같은지 확인한다. 이렇게 하면 코드를 외워 쓰는 것이 아니라 알고리즘의 흐름을 이해한 상태에서 다시 구현할 수 있다.

여기서 중요한 것은 최종 결과만 보는 것이 아니다. 문자열이 리스트가 되고, 리스트의 원소가 정수가 되고, 정수들이 행과 전체 이미지 구조로 쌓이는 중간 과정을 확인해야 한다. 실행 셀에는 의도적으로 변수의 값과 자료형, 길이 또는 계산식을 출력해 어느 단계에서 값이 달라지는지 눈으로 찾을 수 있게 한다.

### 실행 전 예상
1. 셀을 실행하기 전에 어떤 변수가 새로 만들어지는지 적어 본다.
2. `print()`에 표시될 값과 자료형을 먼저 예상한다.
3. 예상과 실제 출력이 다르면 마지막 결과만 고치지 않고 최초로 달라진 중간 변수를 찾는다.

### 실행하면서 확인할 중간 변수
- 입력을 담은 변수: 원본 문자열이나 RGB 튜플이 변하지 않았는지 확인한다.
- 변환 중간값: `split()` 결과, 정수 RGB, 밝기, 누적 합처럼 다음 계산의 입력이 되는 값을 확인한다.
- 크기 정보: 행 수, 열 수, 리스트 길이가 H와 W에 맞는지 확인한다.
- 최종값: 계산식과 조건식이 문제의 기준과 정확히 같은지 확인한다.

### 헷갈린 점과 확인 방법
질문 75개를 따로 외우는 것이 목적이 아니라 하나의 전체 흐름으로 연결하는 것이 목적이다.

헷갈릴 때는 코드를 한꺼번에 바꾸지 않고 현재 값, `type()`, `len()`을 하나씩 출력한다. 이렇게 하면 문법 오류와 알고리즘 오류를 구분할 수 있고, 같은 입력으로 다시 실행했을 때 결과도 재현할 수 있다.

### 직접 실행해 볼 내용
문제 전체 흐름을 함수 세 개로 구성한 작은 완성 예제를 실행한다. 아래 코드 셀을 실행한 뒤 출력 순서를 위의 예상과 대조한다. 값이 맞더라도 왜 맞는지 계산식을 손으로 한 번 따라가고, 가능하면 입력값 하나를 바꿔 결과가 함께 바뀌는지도 확인한다.

### 실행 결과를 읽는 기준
오류 없이 끝났다는 사실만으로 충분하지 않다. 출력된 중간값이 입력 구조와 일치하는지, 정수 나눗셈이 적용되었는지, 행과 열의 개수가 보존되었는지, 경계값에서 `>=` 조건이 의도대로 동작하는지를 확인한다. `assert`가 있는 셀은 아무 메시지 없이 다음 줄로 진행되면 검증을 통과한 것이다.

### 학습 정리
이번 질문의 핵심은 **입력 구조 확인 → 한 픽셀 밝기 계산 → 밝기 맵 구성 → 전체 평균 threshold 계산 → 0/1 변환 → 출력 형식 확인 → 테스트 순서로 접근하면 된다**라는 점이다. 코드를 실행한 뒤에는 이 문장을 보지 않고도 자신의 말로 설명해 본다. 설명이 막히면 출력된 중간 변수 중 어느 값이 왜 만들어졌는지부터 다시 추적한다.

### 다음 질문으로 연결
마지막으로 작은 완성 예제를 처음부터 다시 실행하고 각 단계의 값이 예상과 일치하는지 확인한다.


```python
print('=== Q75: 문제 전체를 처음부터 다시 구성 ===')
def calc_brightness(r, g, b):
    return (r + g + b) // 3

def parse_brightness_map(row_texts, height, width):
    brightness_map = []
    for row_index, row_text in enumerate(row_texts):
        pixel_texts = row_text.split()
        assert len(pixel_texts) == width
        brightness_row = []
        for column_index, pixel_text in enumerate(pixel_texts):
            rgb = tuple(map(int, pixel_text.split(',')))
            assert len(rgb) == 3
            brightness = calc_brightness(*rgb)
            brightness_row.append(brightness)
            print(f'파싱 ({row_index}, {column_index}): {rgb} -> {brightness}')
        brightness_map.append(brightness_row)
    assert len(brightness_map) == height
    return brightness_map

def calc_threshold(brightness_map, height, width):
    total = 0
    for row_index, brightness_row in enumerate(brightness_map):
        row_sum = sum(brightness_row)
        total += row_sum
        print(f'{row_index}행 합={row_sum}, 누적 합={total}')
    threshold = total // (height * width)
    print(f'threshold={total}//{height * width}={threshold}')
    return threshold

def to_black_white(brightness_map, threshold):
    result = []
    for row_index, brightness_row in enumerate(brightness_map):
        result_row = ''.join(
            '1' if brightness >= threshold else '0'
            for brightness in brightness_row
        )
        result.append(result_row)
        print(f'{row_index}행 흑백 결과={result_row}')
    return result

height, width = 3, 3
row_texts = [
    '255,255,255 200,200,200 0,0,0',
    '120,120,120 60,60,60 30,30,30',
    '255,0,0 0,255,0 0,0,255',
]
brightness_map = parse_brightness_map(row_texts, height, width)
threshold = calc_threshold(brightness_map, height, width)
result = to_black_white(brightness_map, threshold)

print('최종 밝기 맵 =', brightness_map)
print('최종 threshold =', threshold)
print('최종 출력:')
print(threshold)
for row in result:
    print(row)

assert threshold == 102
assert result == ['110', '100', '000']
print('[확인] 전체 재구성 및 검증 완료')
```
