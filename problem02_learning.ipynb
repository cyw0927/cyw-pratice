### 왜 필요한가?
최솟값과 최댓값을 알고 있으면 밝기 계산 함수가 정상 범위의 값을 반환하는지 쉽게 검증할 수 있다.

### 직접 해보기

## Q06. 왜 RGB 세 값을 밝기 하나로 바꾸는가?

## Q08. 왜 `/`가 아니라 `//`를 사용하는가?

### 답
소수점 아래를 버린 정수 결과가 필요하기 때문이다.

### 상세 설명
Python에서 `/`는 항상 실수 나눗셈 결과를 만들고, `//`는 몫을 구한다. 예를 들어 `7/3`은 약 2.333이고 `7//3`은 2다. 이 문제의 예시와 조건은 밝기와 threshold를 정수로 다루므로 `//`를 써야 한다.

### 왜 필요한가?
연산자가 달라지면 threshold 값이나 자료형이 달라질 수 있다. 특히 평균이 딱 나누어 떨어지지 않을 때 차이가 생긴다.

### 직접 해보기
`7/3`, `7//3`, `920/9`, `920//9`를 각각 실행해 차이를 비교한다.

### 코드 예시

```python
print(7/3, 7//3)
print(920/9, 920//9)
```

## Q10. 밝기 계산을 함수로 따로 만드는 이유는 무엇인가?

## Q13. 첫 줄 `3 3`을 Python에서 어떻게 두 변수에 저장하는가?

`height, width = map(int, input().split())`처럼 공백 기준으로 나누고 정수로 바꿔 저장한다.

### 상세 설명

### 왜 필요한가?
입력받은 크기를 이후 `range(height)`와 `height * width`에 사용해야 하므로 숫자로 저장해야 한다.

### 직접 해보기

### 왜 필요한가?
