## Q19. `brightness_row`는 왜 매 행마다 새로 만드는가?

### 상세 설명
같은 리스트를 계속 재사용하면 이전 행 값과 다음 행 값이 섞일 수 있다. 각 행마다 새 빈 리스트를 만들고 완료된 행을 전체 리스트에 추가한다.

## Q23. 전체 밝기의 합을 구할 때 왜 `total = 0`으로 시작하는가?

### 답
아직 아무 값도 더하지 않았기 때문에 합의 시작값이 0이기 때문이다.

### 상세 설명
각 밝기를 하나씩 더하려면 누적 변수의 초기값이 필요하다. 덧셈의 항등원인 0에서 시작하면 계산이 자연스럽다.

### 왜 필요한가?
누적합 패턴을 이해하는 기본 단계다.

### 직접 해보기
세 값을 차례로 더하며 total 변화를 확인한다.

```python
values=[255,200,0]
total=0
for v in values:
    total+=v
    print(total)
```

## Q37. `calc_threshold()` 함수를 따로 만드는 이유는 무엇인가?

### 답
threshold 계산 역할을 별도 기능으로 분리하기 위해서다.

### 상세 설명
함수 안에서 전체 밝기 합과 픽셀 수를 이용해 평균을 반환하게 만들면 메인 흐름이 간단해진다.

### 왜 필요한가?
기능별 테스트와 설명이 쉬워진다.

### 직접 해보기
작은 밝기 맵을 함수에 넣어 threshold를 확인한다.

```python
def calc_threshold(m,h,w):
    total=0
    for row in m:
        for b in row:
            total+=b
    return total//(h*w)
print(calc_threshold([[10,20],[30,40]],2,2))
```

## Q38. `to_black_white()` 함수를 따로 만드는 이유는 무엇인가?

### 답

### 상세 설명

### 왜 필요한가?

### 직접 해보기
작은 2x2 밝기 맵을 흑백으로 변환한다.

```python
def to_bw(m,t):
    result=[]
    for row in m:
        s=''
        for b in row:
            s+='1' if b>=t else '0'
        result.append(s)
    return result
print(to_bw([[10,30],[20,40]],25))
```

정상 밝기 범위의 최댓값이 255이므로 255보다 큰 값은 없다. 따라서 흰색처럼 밝기 255인 픽셀만 기준 이상이다.

### 왜 필요한가?
최댓값 경계조건을 이해할 수 있다.

### 직접 해보기
여러 밝기를 threshold 255와 비교한다.

```python
t=255
for b in [0,85,254,255]:
    print(b,1 if b>=t else 0)
```

## Q74. 출력 문자열 길이는 왜 W와 같아야 하는가?

### 답
한 행에 W개의 픽셀이 있으므로 각 픽셀마다 0 또는 1 하나가 나와야 하기 때문이다.

### 상세 설명
결과 행 길이가 W보다 짧거나 길면 일부 픽셀이 빠졌거나 추가된 것이다.

### 왜 필요한가?
출력 구조 검증에 사용할 수 있다.

### 직접 해보기
결과 문자열 길이를 검사한다.

```python
W=3
row='110'
print(len(row), '정상' if len(row)==W else '오류')
```

## Q75. 결과 행 개수는 왜 H와 같아야 하는가?

### 답
입력 이미지에 H개의 행이 있으므로 출력도 H개의 행이 있어야 하기 때문이다.

### 상세 설명
각 입력 행이 하나의 결과 행으로 대응된다.

### 왜 필요한가?
입력과 출력 크기가 보존되는지 확인할 수 있다.

### 직접 해보기
결과 행 개수를 H와 비교한다.

```python
H=3
rows=['110','100','000']
print(len(rows), '정상' if len(rows)==H else '오류')
```
