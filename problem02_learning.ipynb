문제 2를 50개의 실제 이해 단위로 나누어 정리했다. 각 질문은 단순 정의만 적는 것이 아니라 **답 → 상세 설명 → 왜 필요한가 → 직접 해보기 → 코드 예시** 순서로 정리했다.

전체 흐름: `RGB 입력 → 문자열 분리 → 정수 변환 → 밝기 계산 → 2차원 리스트 저장 → 전체 평균(threshold) 계산 → 0/1 변환 → 출력 → 테스트`

## Q01. 문제 2가 근본적으로 원하는 것은 무엇인가?
### 답
컬러 이미지의 RGB 데이터를 각 픽셀의 밝기 값으로 바꾸고, 전체 평균 밝기를 기준으로 각 픽셀을 0 또는 1로 분류하는 것이다.
### 상세 설명
이 문제는 JPG나 PNG 파일을 직접 밝게 만들거나 저장하는 문제가 아니다. 입력으로 들어온 RGB 숫자 데이터를 계산해서 흑백 이진 데이터로 변환하는 알고리즘 문제다. 즉 입력 데이터의 형태가 `R,G,B`이고, 중간에는 brightness라는 정수 값이 되며, 마지막에는 0과 1만 남는다.
### 왜 필요한가?
전체 목표를 먼저 이해해야 뒤에서 배우는 `split`, 반복문, 평균, 조건문이 왜 필요한지 연결해서 설명할 수 있다.
### 직접 해보기
종이에 `RGB → brightness → threshold → 0/1` 순서를 적고 각 단계에서 만들어지는 값을 한 줄씩 설명한다.
### 코드 예시
```python
print('RGB → brightness → threshold → 0/1')
```

## Q02. RGB는 무엇이고 왜 세 개의 값이 필요한가?
### 답
RGB는 Red, Green, Blue의 약자로 픽셀 하나의 색을 나타내는 세 개의 색 성분이다.
### 상세 설명
컴퓨터 화면은 빨강, 초록, 파랑 빛의 세기를 조합해서 색을 표현한다. `(255,0,0)`은 빨강이 가장 강하고, `(0,255,0)`은 초록, `(0,0,255)`는 파랑에 해당한다. 픽셀 하나는 화면에서는 색 하나로 보이지만 프로그램에서는 세 숫자의 묶음으로 다룬다.
### 왜 필요한가?
밝기 계산식을 적용하려면 픽셀 안의 세 성분을 각각 R, G, B로 분리해야 한다.
### 직접 해보기
빨강, 초록, 파랑, 흰색, 검은색의 대표 RGB를 직접 적어 본다.
### 코드 예시
```python
samples = {'red':(255,0,0), 'green':(0,255,0), 'blue':(0,0,255)}
print(samples)
```

## Q03. RGB 값은 왜 0부터 255까지인가?
### 답
일반적인 8비트 색상 성분은 0부터 255까지 총 256단계를 표현할 수 있기 때문이다.
### 상세 설명
8비트는 0과 1을 8자리 사용하는 방식이고 `2**8 = 256`개의 서로 다른 값을 만들 수 있다. 정수 범위로는 0~255다. 0은 해당 색 성분이 없는 상태이고 255는 가장 강한 상태다.
### 왜 필요한가?
입력값의 정상 범위를 알고 있으면 테스트 데이터를 만들기 쉽고, 계산된 밝기도 0~255 사이일 것이라고 예상할 수 있다.
### 직접 해보기
`2**8`과 `2**8-1`을 계산해 본다.
### 코드 예시
```python
print(2**8)
print(2**8 - 1)
```

### 왜 필요한가?
입력 파싱을 정확히 해야 이후 밝기 계산도 정확해진다.

## Q05. 검은색과 흰색의 RGB와 밝기는 각각 얼마인가?
### 답
검은색은 `(0,0,0)`이고 밝기는 0, 흰색은 `(255,255,255)`이고 밝기는 255다.
### 상세 설명
밝기 공식 `(R+G+B)//3`에 넣으면 검정은 0, 흰색은 255가 된다. 두 값은 밝기 계산의 최솟값과 최댓값이므로 함수 테스트에도 매우 유용하다.
### 왜 필요한가?
경계값을 확인하면 계산식이 정상 범위에서 작동하는지 빠르게 검증할 수 있다.
### 직접 해보기
두 RGB 값을 공식에 직접 넣어 계산한다.
### 코드 예시
```python
print((0+0+0)//3)
print((255+255+255)//3)
```

## Q06. 왜 RGB 세 값을 밝기 하나로 바꾸는가?
### 답
흑백 분류에서는 색의 종류보다 밝고 어두운 정도를 하나의 숫자로 비교하는 것이 편하기 때문이다.
### 상세 설명
R,G,B를 그대로 사용하면 픽셀 하나마다 세 값을 동시에 비교해야 한다. brightness 하나로 줄이면 모든 픽셀을 같은 threshold와 비교할 수 있다. 즉 색 정보를 비교하기 쉬운 하나의 점수로 단순화한다.
### 왜 필요한가?
0과 1로 분류하려면 모든 픽셀에 공통적으로 적용할 수 있는 대표 숫자가 필요하다.
### 직접 해보기
빨강 `(255,0,0)`과 회색 `(85,85,85)`의 밝기가 둘 다 85인지 확인한다.
### 코드 예시
```python
for rgb in [(255,0,0),(85,85,85)]:
    print(sum(rgb)//3)
```

## Q07. 이 문제에서 밝기는 정확히 어떻게 계산하는가?
### 답
`brightness = (R + G + B) // 3`으로 계산한다.
### 상세 설명
R,G,B를 같은 비중으로 더하고 3으로 나눈 정수 평균을 사용한다. 실제 영상 처리에는 가중치가 있는 밝기 공식도 있지만 이 과제에서는 문제에 적힌 공식을 그대로 사용해야 한다.
### 왜 필요한가?
문제의 핵심 계산식이므로 여기서 값이 틀리면 threshold와 최종 결과가 모두 틀어진다.
### 직접 해보기
`(120,120,120)`과 `(255,0,0)`을 직접 계산한다.
### 코드 예시
```python
print((120+120+120)//3)
print((255+0+0)//3)
```

## Q08. 왜 `/`가 아니라 `//`를 사용하는가?
### 답
이 문제는 소수점 아래를 버린 정수 밝기를 사용하기 때문이다.
### 상세 설명
Python에서 `/`는 실수 나눗셈이고 `//`는 몫을 구하는 정수 나눗셈이다. `7/3`은 약 2.333이고 `7//3`은 2다. 문제 예시도 정수 나눗셈 기준으로 작성되어 있다.
### 왜 필요한가?
연산자를 다르게 쓰면 threshold 값과 비교 결과가 달라질 수 있다.
### 직접 해보기
두 나눗셈 결과와 자료형을 비교한다.
### 코드 예시
```python
print(7/3, type(7/3))
print(7//3, type(7//3))
```

## Q09. 빨간색 `(255,0,0)`의 밝기는 왜 85인가?
### 답
`(255+0+0)//3 = 85`이기 때문이다.
### 상세 설명
빨강 성분 하나가 매우 커도 세 성분을 같은 비중으로 평균내기 때문에 결과는 85다. 색이 강하다는 것과 이 문제에서 정의한 brightness가 255라는 것은 같은 의미가 아니다.
### 왜 필요한가?
RGB 값과 밝기 값이 서로 다른 정보라는 점을 이해할 수 있다.

초록과 파랑도 각각 85인지 확인한다.

```python
for rgb in [(255,0,0),(0,255,0),(0,0,255)]:
    print(sum(rgb)//3)
```

## Q10. 밝기 계산을 함수로 분리하는 이유는 무엇인가?
### 답
같은 계산을 재사용하고 코드의 역할을 명확하게 만들기 위해서다.
### 상세 설명
`calc_brightness()`가 한 픽셀의 밝기만 담당하면 메인 코드가 단순해지고 함수만 따로 테스트할 수 있다. 나중에 공식을 수정할 때도 한 함수만 바꾸면 된다.
### 왜 필요한가?
기능을 나누면 읽기, 테스트, 디버깅, 설명이 모두 쉬워진다.
### 직접 해보기
200,200,200을 넣어 200이 나오는지 확인한다.
### 코드 예시
```python
def calc_brightness(r,g,b):
    return (r+g+b)//3
print(calc_brightness(200,200,200))
```

## Q12. 전체 픽셀 수는 왜 `H * W`인가?
### 답
H개의 행마다 W개의 픽셀이 있기 때문이다.
### 상세 설명
3행 4열이면 한 행마다 4픽셀이 있고 그런 행이 3개이므로 12픽셀이다. 평균을 구할 때 전체 합을 이 픽셀 개수로 나눈다.
### 왜 필요한가?
threshold 계산의 분모가 된다.
### 직접 해보기
2×5, 3×3, 4×4의 픽셀 수를 계산한다.
### 코드 예시
```python
for h,w in [(2,5),(3,3),(4,4)]:
    print(h*w)
```

## Q13. 첫 줄 `3 3`은 어떻게 두 변수에 저장하는가?
### 답
공백 기준으로 나누고 정수로 변환해 `height, width`에 저장한다.
### 상세 설명
`input()` 결과는 문자열이고 `split()`이 두 문자열로 나눈다. 그다음 `map(int, ...)`가 정수로 바꾸고 두 변수에 각각 저장한다.
### 왜 필요한가?
입력받은 이미지 크기를 반복문과 threshold 계산에서 사용해야 한다.
### 직접 해보기
문자열 `'3 4'`에 같은 과정을 적용한다.
### 코드 예시
```python
text='3 4'
height,width=map(int,text.split())
print(height,width)
```

## Q17. `map(int, ...)`는 왜 사용하는가?
### 답
여러 문자열을 한 번에 정수로 변환하기 편하기 때문이다.
### 상세 설명
`pixel.split(',')`은 세 문자열을 만든다. 각각 `int()`를 쓰는 대신 `map(int, ...)`으로 세 값 모두에 같은 변환을 적용하고 `r,g,b` 세 변수에 나눠 넣을 수 있다.
### 왜 필요한가?
입력 처리 코드를 간결하고 읽기 쉽게 만들 수 있다.

세 값을 변환하고 type을 확인한다.
### 코드 예시
```python
r,g,b=map(int,'255,0,0'.split(','))
print(r,g,b,type(r))
```

## Q18. 왜 중첩 반복문이 필요한가?
### 답
이미지가 행과 열로 된 2차원 데이터이기 때문이다.
### 상세 설명
바깥 반복문은 행을 하나씩 처리하고 안쪽 반복문은 현재 행의 픽셀을 하나씩 처리한다. 이렇게 하면 H×W개의 모든 픽셀을 빠짐없이 방문할 수 있다.
### 왜 필요한가?
모든 픽셀에 동일한 밝기 계산을 적용하기 위해 필요하다.
### 직접 해보기
2×2 리스트를 중첩 반복문으로 순회한다.
### 코드 예시
```python
data=[[1,2],[3,4]]
for row in data:
    for x in row:
        print(x)
```

## Q19. `brightness_row`는 왜 매 행마다 새로 만드는가?
### 답
현재 행의 밝기만 따로 모아서 원래 이미지의 행 구조를 유지하기 위해서다.
### 상세 설명
한 리스트를 계속 재사용하면 이전 행과 다음 행의 값이 섞인다. 그래서 새로운 행이 시작될 때 빈 리스트를 만들고 한 행을 모두 계산한 다음 전체 밝기 맵에 추가한다.
### 왜 필요한가?
2차원 구조를 유지해야 최종 결과도 같은 행 단위로 출력할 수 있다.
### 직접 해보기
1행과 2행을 따로 리스트로 만든 뒤 전체 리스트에 넣어 본다.
### 코드 예시
```python
brightness_map=[]
for row in [[10,20],[30,40]]:
    current=[]
    for x in row:
        current.append(x)
    brightness_map.append(current)
print(brightness_map)
```

## Q20. `append()`는 어떤 역할을 하는가?
### 답

## Q33. `calc_threshold()` 함수를 따로 만드는 이유는 무엇인가?
### 답
전체 밝기 평균을 구하는 역할을 하나의 함수로 분리하기 위해서다.
### 상세 설명
threshold 계산에는 전체 합 구하기, 픽셀 개수 구하기, 정수 평균 계산하기가 포함된다. 이 세 작업을 하나의 함수에 묶으면 main에서는 `threshold = calc_threshold(...)`라는 한 줄로 의미를 읽을 수 있다. 또한 threshold가 이상할 때 이 함수만 따로 확인할 수 있다.
### 왜 필요한가?
기능별로 책임을 나누면 디버깅과 설명이 쉬워진다.
### 직접 해보기
작은 2×2 밝기 맵으로 예상 평균과 함수 결과를 비교한다.
### 코드 예시
```python
def calc_threshold(m,h,w):
    total=0
    for row in m:
        for v in row:
            total+=v
    return total//(h*w)
print(calc_threshold([[10,20],[30,40]],2,2))
```

```python
pixels=[(255,255,255),(200,200,200),(0,0,0),(120,120,120),(60,60,60),(30,30,30),(255,0,0),(0,255,0),(0,0,255)]
print([sum(p)//3 for p in pixels])
```

## Q38. 예제에서 전체 밝기 합 920은 어떻게 나오는가?
### 답
9개 밝기 `255+200+0+120+60+30+85+85+85`를 모두 더하면 920이 된다.
### 상세 설명
threshold는 평균이기 때문에 전체 합이 정확해야 한다. 중간 합이 하나라도 틀리면 threshold가 달라지고 그 결과 모든 픽셀의 0/1 판정까지 연쇄적으로 달라질 수 있다. 따라서 예제에서는 920이라는 중간값을 직접 검증하는 것이 좋다.
### 왜 필요한가?
오류 위치를 단계별로 찾는 습관을 기르기 위해 중요하다.
### 직접 해보기
손으로 더한 결과와 Python `sum()` 결과를 비교한다.
### 코드 예시

## Q39. 예제에서 threshold 102는 어떻게 나오는가?
### 답
전체 합 920을 픽셀 수 9로 정수 나눗셈해 `920 // 9 = 102`가 된다.
### 상세 설명
이미지는 3×3이므로 픽셀은 9개다. 일반 나눗셈이면 약 102.22지만 이 문제는 `//`를 사용하므로 소수점 아래를 버려 102가 된다. 이 102가 이후 9개 픽셀 모두에게 적용되는 공통 기준선이다.
### 왜 필요한가?
예제 핵심 값이 어떤 계산 과정을 거쳐 나오는지 설명할 수 있어야 한다.

`920/9`와 `920//9`를 비교한다.

## Q40. 예제 첫 번째 행이 왜 `110`이 되는가?
### 답
밝기 `[255,200,0]`을 threshold 102와 비교하면 각각 1,1,0이 되기 때문이다.
### 상세 설명
255와 200은 102 이상이므로 1이고, 0은 102 미만이므로 0이다. 세 결과를 한 행 문자열로 이어 붙이면 `110`이 된다. 이 과정은 코드의 if/else가 실제 출력 한 줄을 만드는 방식을 그대로 보여준다.
### 왜 필요한가?
조건 비교와 문자열 생성이 연결되는 과정을 직접 추적할 수 있다.
### 직접 해보기
종이에 `255→1`, `200→1`, `0→0`이라고 적는다.
### 코드 예시
```python
th=102
row=[255,200,0]
print(''.join('1' if v>=th else '0' for v in row))
```

## Q41. 예제 두 번째 행이 왜 `100`이 되는가?

120은 기준 이상이라 1, 60과 30은 기준 미만이라 0이다. 따라서 결과는 `100`이다. 첫 번째 행과 완전히 같은 규칙을 다른 데이터에 적용한 것이다.
### 왜 필요한가?
한 번 정한 규칙이 모든 행과 픽셀에 동일하게 적용된다는 점을 확인한다.

각 값을 102와 비교해 True/False를 적는다.
### 코드 예시

## Q42. 예제 세 번째 행이 왜 `000`이 되는가?

## Q43. 전체가 검은색인 이미지에서 왜 결과가 모두 `1`이 될 수 있는가?
### 답
모든 brightness가 0이면 threshold도 0이고 `0 >= 0`이 참이기 때문이다.
### 상세 설명
직관적으로 검은색이면 0이라고 생각하기 쉽지만 이 과제의 규칙은 'threshold 이상이면 1'이다. 전체 픽셀이 검정이면 평균도 0이고 모든 픽셀은 평균과 정확히 같다. 따라서 문제 규칙을 그대로 따르면 전부 1이 된다. 이것은 버그가 아니라 threshold와 `>=` 조건이 만들어 내는 경계 사례다.
### 왜 필요한가?
직관보다 문제에서 주어진 규칙을 우선해야 한다는 것을 이해할 수 있다.
### 직접 해보기
1×1 검정과 2×2 검정을 손으로 계산한다.
### 코드 예시
```python
b=0
th=0
print('1' if b>=th else '0')
```

## Q44. 전체가 흰색인 이미지에서는 왜 결과가 모두 `1`인가?
### 답
모든 brightness가 255이고 threshold도 255라서 모두 `255 >= 255`를 만족하기 때문이다.
### 상세 설명
같은 값들만 평균내면 평균도 같은 값이다. 따라서 흰색만 있는 이미지의 threshold는 255다. 각 픽셀은 threshold보다 크지는 않지만 정확히 같기 때문에 `>=` 조건에서 참이 된다.
### 왜 필요한가?
최댓값 경계에서도 `>=`가 어떻게 작동하는지 확인할 수 있다.
### 직접 해보기
2×2 흰색 이미지의 합과 평균을 직접 계산한다.
### 코드 예시
```python
values=[255,255,255,255]
th=sum(values)//len(values)
print(th,[v>=th for v in values])
```

## Q45. 1×1 이미지에서는 왜 결과가 항상 `1`인가?
### 답
픽셀이 하나뿐이면 그 픽셀의 brightness가 그대로 threshold가 되기 때문이다.
### 상세 설명
예를 들어 밝기가 30인 픽셀 하나만 있다면 전체 합은 30, 픽셀 수는 1, threshold는 30이다. 비교하면 `30 >= 30`이므로 결과는 1이다. 검정이든 회색이든 흰색이든 픽셀이 하나뿐이면 자기 자신이 평균이므로 같은 구조가 된다.
### 왜 필요한가?
가장 작은 입력 크기에서 알고리즘이 어떻게 작동하는지 검증하는 경계 테스트다.
### 직접 해보기
밝기 0,30,255 각각에 대해 결과를 계산한다.
### 코드 예시
```python
for b in [0,30,255]:
    th=b
    print(b,'1' if b>=th else '0')
```

## Q46. 입력 픽셀 개수가 W와 다르면 어떤 문제가 생길 수 있는가?
### 답
실제 저장된 픽셀 수와 `H*W`로 계산한 픽셀 수가 달라져 threshold가 잘못될 수 있다.
### 상세 설명
현재 과제 코드는 정상 입력이 들어온다고 가정한다. 하지만 실제 프로그램에서 W=3인데 한 행에 두 픽셀만 입력되면 brightness_map에는 실제로 더 적은 값이 들어간다. 그런데 분모는 여전히 H×W를 쓰면 평균이 잘못된다. 과제에서는 필수 예외 처리가 아니지만 실제 프로그램에서는 입력 길이 검증이 필요하다.
### 왜 필요한가?
알고리즘 문제의 정상 입력 가정과 실제 프로그램의 방어적 코딩 차이를 이해할 수 있다.
### 직접 해보기
W=3인데 두 픽셀만 들어온 경우 expected와 actual을 비교한다.
### 코드 예시

## Q47. 이 알고리즘의 시간 복잡도는 어떻게 생각할 수 있는가?
### 답
모든 픽셀을 상수 횟수만큼 순회하므로 전체적으로 `O(H×W)`라고 볼 수 있다.
### 상세 설명
밝기를 만들 때 H×W개 픽셀을 보고, 전체 합을 구할 때 다시 보고, 0/1 변환 때 다시 본다. 전체를 약 세 번 순회하지만 Big-O에서는 3 같은 상수배를 생략하므로 `O(H×W)`다. 즉 픽셀 수가 두 배가 되면 해야 할 일도 대략 두 배가 된다.
### 왜 필요한가?
이미지 크기가 커질 때 실행 시간이 어떻게 증가하는지 설명하기 위해 필요하다.
### 직접 해보기
10×10과 100×100의 픽셀 수를 비교한다.
### 코드 예시
```python
for h,w in [(10,10),(100,100)]:
    print(h*w)
```

## Q48. 공간 복잡도는 어떻게 생각할 수 있는가?
### 답
밝기 맵과 결과를 이미지 크기만큼 저장하므로 대략 `O(H×W)`의 추가 공간을 사용한다.
### 상세 설명
현재 구현은 모든 brightness를 2차원 리스트에 저장하고, 변환된 흑백 결과도 별도로 저장한다. 따라서 픽셀이 많아질수록 메모리에 저장되는 값도 비례해서 늘어난다. 입력하면서 합계를 동시에 계산하면 반복 횟수 일부는 줄일 수 있지만 threshold가 결정되기 전에 개별 분류를 할 수 없어서 중간 데이터를 다시 사용할 방법은 필요하다.
### 왜 필요한가?
실행 시간뿐 아니라 메모리 사용량도 입력 크기의 영향을 받는다는 점을 이해할 수 있다.
### 직접 해보기
3×3과 100×100에서 저장해야 할 brightness 개수를 비교한다.
### 코드 예시
```python
for h,w in [(3,3),(100,100)]:
    print(h*w)
```

## Q49. 다른 해결 방법에는 무엇이 있을 수 있는가?
### 답
1차원 리스트 사용, 입력과 동시에 total 누적, 라이브러리를 이용한 배열 연산 등 여러 방법이 있다.

### 왜 필요한가?
한 문제에 하나의 정답 방식만 있는 것이 아니라는 점과 내가 선택한 방식의 장단점을 설명할 수 있게 된다.

### 코드 예시

## Q50. 50개 질문을 모두 연결하면 문제 2의 전체 흐름은 무엇인가?

50개의 작은 질문을 다시 하나의 알고리즘으로 합쳐서 설명할 수 있어야 실제로 문제를 이해했다고 볼 수 있다.
