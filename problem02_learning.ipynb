# 문제 2 학습 노트 — 50개 질문 상세판

문제 2를 50개의 실제 이해 단위로 나누어 정리했다. 각 항목은 **답 → 상세 설명 → 왜 필요한가 → 직접 확인/예시** 순서로 작성했다.

전체 흐름: `RGB 입력 → 밝기 계산 → 전체 평균(threshold) → threshold와 비교 → 0/1 변환 → 출력 → 테스트`

---

## Q01. 문제 2가 근본적으로 원하는 것은 무엇인가?
### 답
컬러 이미지의 RGB 데이터를 밝기 값으로 바꾸고, 전체 평균 밝기를 기준으로 각 픽셀을 0 또는 1로 분류하는 것이다.
### 상세 설명
실제 JPG나 PNG 파일을 밝게 보정하는 문제가 아니라, 주어진 RGB 숫자 데이터를 계산해서 흑백 이진 데이터로 바꾸는 알고리즘 문제다.
### 왜 필요한가?
문제의 목표를 먼저 이해해야 이후의 입력 처리, 평균 계산, 조건 비교가 하나의 흐름으로 연결된다.
### 직접 확인
`RGB → brightness → threshold → 0/1` 순서로 적어 본다.

## Q02. RGB는 무엇인가?
### 답
RGB는 Red, Green, Blue의 약자로 픽셀 하나의 색을 나타내는 세 값이다.
### 상세 설명
컴퓨터 화면은 빨강, 초록, 파랑 빛의 세기를 조합해 색을 표현한다. 예를 들어 `(255,0,0)`은 빨강 성분이 가장 강한 픽셀이다.
### 왜 필요한가?
입력값이 RGB 형식이므로 세 값을 분리해야 밝기 공식을 적용할 수 있다.
### 예시
`(255,0,0)`은 빨강, `(0,255,0)`은 초록, `(0,0,255)`는 파랑이다.

## Q03. 하나의 픽셀은 몇 개의 숫자로 구성되는가?
### 답
이 문제에서는 R, G, B 세 개의 정수로 구성된다.
### 상세 설명
`255,0,0`은 하나의 문자열처럼 보이지만 실제 의미는 R=255, G=0, B=0이라는 세 개의 값이다.
### 왜 필요한가?
픽셀 단위와 색 성분 단위를 구분해야 입력을 정확히 나눌 수 있다.
### 예시
`'255,0,0'.split(',')`을 생각해 보면 세 조각으로 나뉜다.

## Q04. RGB 값의 범위는 왜 0부터 255인가?
### 답
일반적인 8비트 색상 성분은 0부터 255까지 총 256단계를 표현하기 때문이다.
### 상세 설명
0은 해당 색 성분이 없는 상태이고 255는 가장 강한 상태다. 이 문제도 이 범위의 RGB 정수를 사용한다.
### 왜 필요한가?
정상 입력의 범위를 알고 있어야 테스트값을 만들고 결과 범위를 예상할 수 있다.
### 예시
`2**8 = 256`, 따라서 값은 0~255다.

## Q05. 검은색과 흰색의 RGB는 무엇인가?
### 답
검은색은 `(0,0,0)`, 흰색은 `(255,255,255)`이다.
### 상세 설명
세 색 성분이 모두 0이면 빛이 없는 상태이고, 모두 최대이면 흰색이 된다.
### 왜 필요한가?
밝기 계산의 최소값과 최대값을 확인하는 대표 테스트가 된다.
### 직접 확인
검정 밝기는 0, 흰색 밝기는 255가 된다.

## Q06. 왜 RGB를 밝기 하나의 숫자로 바꾸는가?
### 답
흑백 분류에서는 색의 종류보다 얼마나 밝고 어두운지가 중요하기 때문이다.
### 상세 설명
R, G, B 세 숫자를 동시에 비교하는 대신 하나의 대표 밝기 값으로 바꾸면 모든 픽셀을 같은 기준으로 비교할 수 있다.
### 왜 필요한가?
나중에 모든 픽셀을 하나의 threshold와 비교하려면 픽셀마다 대표 숫자 하나가 필요하다.
### 예시
`(255,0,0)`은 이 문제의 공식으로 밝기 85가 된다.

## Q07. 밝기는 어떻게 계산하는가?
### 답
`brightness = (R + G + B) // 3`으로 계산한다.
### 상세 설명
세 색 성분을 모두 더한 뒤 3으로 나눈 정수 평균을 밝기로 사용한다. 과제에서 지정한 공식이므로 그대로 사용한다.
### 왜 필요한가?
문제 2의 가장 기본이 되는 계산식이다.
### 예시
`(120+120+120)//3 = 120`.

## Q08. 왜 `/`가 아니라 `//`를 사용하는가?
### 답
소수점 아래를 버린 정수 결과가 필요하기 때문이다.
### 상세 설명
Python의 `/`는 실수 나눗셈이고 `//`는 몫을 구하는 정수 나눗셈이다. 문제의 밝기와 threshold는 정수로 계산한다.
### 왜 필요한가?
연산자를 잘못 쓰면 예제와 다른 값이 나오거나 자료형이 달라질 수 있다.
### 예시
`7/3`은 약 2.333, `7//3`은 2다.

## Q09. 빨간색 `(255,0,0)`의 밝기는 얼마인가?
### 답
85이다.
### 상세 설명
공식에 넣으면 `(255+0+0)//3 = 85`가 된다. 색이 강하다고 해서 항상 밝기 255가 되는 것은 아니다.
### 왜 필요한가?
RGB와 밝기가 같은 개념이 아니라는 점을 이해할 수 있다.
### 직접 확인
초록과 파랑도 같은 방식이면 각각 85가 된다.

## Q10. 밝기 계산을 함수로 분리하는 이유는 무엇인가?
### 답
같은 계산을 재사용하고 코드의 역할을 명확하게 만들기 위해서다.
### 상세 설명
`calc_brightness(r,g,b)`가 픽셀 하나의 밝기 계산만 담당하면 메인 코드가 단순해지고 개별 테스트도 쉬워진다.
### 왜 필요한가?
기능별로 오류를 찾고 설명하기 쉬워진다.
### 예시
`calc_brightness(200,200,200)`의 기대값은 200이다.

## Q11. H와 W는 각각 무엇을 뜻하는가?
### 답
H는 이미지 높이, W는 이미지 너비다.
### 상세 설명
H는 행의 개수이고 W는 한 행에 들어 있는 픽셀의 개수다. 입력이 `3 3`이면 3행 3열 이미지다.
### 왜 필요한가?
몇 줄을 읽을지와 전체 픽셀 수를 계산할 때 필요하다.
### 예시
`H=3, W=3`이면 총 9픽셀이다.

## Q12. 전체 픽셀 수는 왜 `H * W`인가?
### 답
H개의 행마다 W개의 픽셀이 있기 때문이다.
### 상세 설명
3행 4열이면 한 줄에 4개씩 총 3줄이므로 12개다. 평균을 구하려면 전체 합과 전체 개수가 모두 필요하다.
### 왜 필요한가?
threshold 계산의 분모가 된다.
### 예시
`3 * 4 = 12`.

## Q13. 첫 줄 `3 3`은 어떻게 두 변수에 저장하는가?
### 답
공백 기준으로 나눈 뒤 정수로 바꿔 `height, width`에 저장한다.
### 상세 설명
`input().split()`은 두 문자열을 만들고 `map(int, ...)`가 두 값을 정수로 바꾼다.
### 왜 필요한가?
이미지 크기를 이후 반복문과 평균 계산에 사용해야 한다.
### 예시
`height, width = map(int, input().split())`.

## Q14. 한 줄의 여러 RGB 픽셀은 어떻게 나누는가?
### 답
공백 기준 `split()`으로 나눈다.
### 상세 설명
한 줄이 `255,0,0 0,255,0`처럼 되어 있으므로 먼저 공백을 기준으로 픽셀 단위로 분리한다.
### 왜 필요한가?
입력 파싱의 첫 번째 단계다.
### 예시
`line.split()`은 두 개의 픽셀 문자열을 만든다.

## Q15. RGB 문자열 `255,0,0`은 어떻게 나누는가?
### 답
`split(',')`으로 쉼표 기준 세 조각으로 나눈다.
### 상세 설명
공백으로 픽셀을 나눈 뒤에도 픽셀 하나는 문자열이므로 내부의 R,G,B를 다시 쉼표 기준으로 분리해야 한다.
### 왜 필요한가?
R,G,B 각각을 숫자로 꺼내기 위한 두 번째 파싱 단계다.
### 예시
`'255,0,0'.split(',')` → `['255','0','0']`.

## Q16. 문자열 숫자는 왜 `int()`로 바꾸는가?
### 답
문자열 상태로는 원하는 숫자 계산을 할 수 없기 때문이다.
### 상세 설명
입력받은 `'255'`는 문자다. 밝기 계산을 하려면 정수 255로 변환해야 한다.
### 왜 필요한가?
문자열 결합이 아니라 실제 산술 연산을 해야 하기 때문이다.
### 예시
`int('255') + 1 = 256`.

## Q17. `map(int, ...)`는 왜 사용하는가?
### 답
여러 문자열을 한 번에 정수로 변환하기 편하기 때문이다.
### 상세 설명
`r, g, b = map(int, pixel.split(','))`처럼 쓰면 세 문자열에 각각 `int()`를 적용할 수 있다.
### 왜 필요한가?
입력 처리 코드를 짧고 명확하게 만든다.
### 예시
`'255,0,0'`이 정수 `255, 0, 0`으로 변환된다.

## Q18. 왜 2차원 리스트에 밝기를 저장하는가?
### 답
이미지의 행과 열 구조를 그대로 유지하기 쉽기 때문이다.
### 상세 설명
이미지는 표처럼 생겼으므로 밝기도 `[[255,200,0],[120,60,30]]`처럼 저장하면 위치를 쉽게 이해할 수 있다.
### 왜 필요한가?
최종 결과도 행 단위로 출력해야 하므로 구조를 보존하는 편이 편하다.
### 예시
`brightness_map[1][0]`은 두 번째 행 첫 번째 밝기다.

## Q19. 1차원 리스트를 사용해도 되는가?
### 답
가능하지만 이번 문제에서는 2차원 구조가 더 직관적이다.
### 상세 설명
합과 평균만 구한다면 모든 밝기를 한 줄로 저장해도 된다. 하지만 행 단위 출력에서는 다시 위치를 계산해야 한다.
### 왜 필요한가?
자료구조 선택이 코드 복잡도와 가독성에 영향을 준다는 점을 이해할 수 있다.
### 예시
`[255,200,0,120,60,30]`도 합 계산은 가능하다.

## Q20. 중첩 반복문은 왜 필요한가?
### 답
모든 행과 각 행 안의 모든 픽셀을 차례로 처리해야 하기 때문이다.
### 상세 설명
바깥 반복문은 행을 이동하고 안쪽 반복문은 현재 행의 픽셀을 이동한다. 2차원 이미지 구조와 자연스럽게 대응된다.
### 왜 필요한가?
모든 픽셀을 빠짐없이 처리할 수 있다.
### 예시
`for row in image:` 안에서 다시 `for pixel in row:`를 사용한다.

## Q21. `brightness_row`는 왜 매 행마다 새로 만드는가?
### 답
현재 행의 밝기만 따로 모으기 위해서다.
### 상세 설명
새 행에서도 이전 리스트를 계속 쓰면 값이 한 곳에 섞인다. 따라서 행을 시작할 때 빈 리스트를 새로 만든다.
### 왜 필요한가?
원래 이미지의 행 구조를 유지한다.
### 예시
1행 `[255,200,0]`, 2행 `[120,60,30]`을 별도 리스트로 만든다.

## Q22. `append()`는 어떤 역할을 하는가?
### 답
리스트 끝에 새로운 값을 추가한다.
### 상세 설명
계산한 밝기를 현재 행에 넣을 때도 쓰고, 완성된 행을 전체 밝기 맵에 넣을 때도 사용한다.
### 왜 필요한가?
반복하면서 계산되는 값을 순서대로 저장할 수 있다.
### 예시
`brightness_row.append(85)`.

## Q23. 전체 밝기 합은 어떻게 구하는가?
### 답
모든 밝기를 반복하면서 `total`에 계속 더한다.
### 상세 설명
`total = 0`에서 시작해 각 픽셀의 brightness를 누적하면 마지막에 전체 합이 된다.
### 왜 필요한가?
전체 평균인 threshold를 계산하려면 합이 필요하다.
### 예시
예제의 밝기 합은 `920`이다.

## Q24. threshold는 어떻게 계산하는가?
### 답
`전체 밝기 합 // 전체 픽셀 수`로 계산한다.
### 상세 설명
이 문제에서는 전체 이미지의 평균 밝기를 밝고 어두움을 나누는 기준값으로 사용한다.
### 왜 필요한가?
모든 픽셀을 0 또는 1로 분류할 공통 기준이 필요하다.
### 예시
`920 // 9 = 102`.

## Q25. 왜 threshold를 평균으로 정하는가?
### 답
과제에서 전체 평균 밝기를 기준선으로 사용하도록 정했기 때문이다.
### 상세 설명
평균은 이미지 전체의 대략적인 밝기 수준을 나타낸다. 평균보다 밝은 픽셀과 어두운 픽셀을 나누는 기준으로 쓸 수 있다.
### 왜 필요한가?
threshold의 의미를 알아야 결과 0과 1을 해석할 수 있다.
### 예시
반 평균을 기준으로 평균 이상과 미만을 나누는 것과 비슷하다.

## Q26. threshold와 같은 밝기는 0인가 1인가?
### 답
1이다.
### 상세 설명
문제 조건이 `brightness >= threshold`이므로 정확히 같은 값도 '이상'에 포함된다.
### 왜 필요한가?
경계값에서 결과가 달라지므로 조건을 정확히 구현해야 한다.
### 예시
`102 >= 102`는 참이다.

## Q27. `>=`와 `>`는 어떤 차이가 있는가?
### 답
`>=`는 같은 값까지 포함하고 `>`는 같은 값을 제외한다.
### 상세 설명
threshold와 정확히 같은 픽셀을 1로 만들려면 반드시 `>=`를 사용해야 한다.
### 왜 필요한가?
연산자 하나만 달라도 테스트 결과가 달라진다.
### 예시
80점 이상에는 80점이 포함되지만 80점 초과에는 포함되지 않는다.

## Q28. 0과 1은 무엇을 뜻하는가?
### 답
이 문제에서는 threshold 이상을 1, 미만을 0으로 표현한다.
### 상세 설명
각 픽셀의 복잡한 RGB 값이 최종적으로 두 종류의 값으로 단순화된다.
### 왜 필요한가?
최종 출력 형식을 이해하기 위해 필요하다.
### 예시
밝기 200, threshold 102라면 결과는 1이다.

## Q29. 왜 결과를 문자열로 이어 붙이는가?
### 답
한 행을 `110`처럼 공백 없이 출력하기 쉽기 때문이다.
### 상세 설명
숫자 리스트 `[1,1,0]`보다 문자열 `'110'`을 만들면 문제의 출력 형식과 바로 맞는다.
### 왜 필요한가?
출력 형식을 정확하게 맞출 수 있다.
### 예시
`result_row += '1'`.

## Q30. 왜 threshold 계산 후 밝기 맵을 다시 순회하는가?
### 답
threshold를 알아야 각 픽셀을 0 또는 1로 결정할 수 있기 때문이다.
### 상세 설명
첫 번째 과정에서는 모든 밝기를 계산하고 평균을 구한다. 그 평균이 나온 뒤에야 각 픽셀을 기준과 비교할 수 있다.
### 왜 필요한가?
알고리즘이 두 번 순회하는 이유를 이해할 수 있다.
### 예시
1차 순회: 밝기 저장, 2차 순회: 0/1 변환.

## Q31. 한 번만 순회해서 처리할 수 없는가?
### 답
입력을 저장하지 않고 즉시 최종 0/1을 정하는 것은 어렵다.
### 상세 설명
threshold는 모든 픽셀의 밝기를 본 뒤에야 알 수 있기 때문에 앞쪽 픽셀을 읽는 시점에는 최종 기준을 아직 모른다.
### 왜 필요한가?
왜 중간값을 저장하는 설계가 필요한지 이해할 수 있다.
### 예시
첫 픽셀을 읽었을 때는 전체 평균을 아직 알 수 없다.

## Q32. `calc_threshold()`를 함수로 분리하는 이유는 무엇인가?
### 답
전체 평균 계산이라는 역할을 독립적으로 관리하기 위해서다.
### 상세 설명

### 왜 필요한가?

### 예시
`calc_threshold([[0,100]], 1, 2)`의 결과를 따로 확인할 수 있다.

## Q33. `to_black_white()`를 함수로 분리하는 이유는 무엇인가?

## Q36. 예제의 brightness_map은 어떻게 되는가?
### 답
`[[255,200,0],[120,60,30],[85,85,85]]`가 된다.
### 상세 설명

## Q37. 예제의 threshold는 왜 102인가?
### 답
전체 밝기 합 920을 9픽셀로 정수 나눗셈했기 때문이다.
### 상세 설명

### 왜 필요한가?
예제 출력의 근거를 직접 계산해 확인할 수 있다.
### 직접 확인
`920 // 9 = 102`.

## Q38. 첫 번째 행이 왜 `110`이 되는가?
### 답
밝기 255와 200은 102 이상이고, 0은 102 미만이기 때문이다.
### 상세 설명
각 픽셀을 순서대로 비교하면 `1,1,0`이 되고 이를 문자열로 붙이면 `110`이다.
### 왜 필요한가?
조건 비교가 실제 출력으로 어떻게 이어지는지 확인할 수 있다.
### 직접 확인
`255>=102`, `200>=102`, `0<102`.

## Q39. 두 번째 행이 왜 `100`이 되는가?
### 답
120만 threshold 이상이고 60과 30은 미만이기 때문이다.
### 상세 설명
세 값을 순서대로 비교하면 `1,0,0`이 된다.
### 왜 필요한가?
여러 값이 섞인 행에서 조건문이 올바르게 동작하는지 확인할 수 있다.
### 직접 확인
`120>=102`, `60<102`, `30<102`.

## Q40. 세 번째 행이 왜 `000`이 되는가?
### 답
세 픽셀의 밝기가 모두 85로 threshold 102보다 작기 때문이다.
### 상세 설명
빨강, 초록, 파랑은 색은 다르지만 이 문제의 단순 평균 밝기 공식에서는 모두 85가 된다.
### 왜 필요한가?
색이 달라도 밝기 값이 같을 수 있음을 확인할 수 있다.
### 직접 확인
`85<102`가 세 번 반복된다.

## Q41. 전체가 검은색이면 결과는 어떻게 되는가?
### 답
과제 규칙대로라면 threshold가 0이고 모든 픽셀이 `0 >= 0`을 만족하므로 모두 1이 된다.
### 상세 설명
직관적으로 검정이 0일 것 같지만 이 문제는 색 자체가 아니라 threshold 이상 여부를 1로 정의한다.
### 왜 필요한가?
경계값 테스트가 왜 중요한지 보여 주는 사례다.
### 직접 확인
`brightness=0`, `threshold=0`, 따라서 결과는 1.

## Q42. 전체가 흰색이면 결과는 어떻게 되는가?
### 답
threshold가 255이고 모든 밝기도 255이므로 모두 1이 된다.
### 상세 설명
같은 값도 `>=` 조건에 포함되기 때문에 전부 1이다.
### 왜 필요한가?
최대 경계값에서도 비교 규칙이 올바른지 확인할 수 있다.
### 직접 확인
`255 >= 255`는 참이다.

## Q43. 1×1 이미지에서는 어떻게 되는가?
### 답
그 픽셀의 밝기가 곧 threshold가 되므로 결과는 항상 1이다.
### 상세 설명
픽셀이 하나뿐이면 전체 평균은 그 픽셀 값 자체다. 따라서 brightness와 threshold가 같다.
### 왜 필요한가?
가장 작은 입력 크기를 테스트하는 경계 사례다.
### 예시
`1 1`에 `30,30,30`을 넣으면 threshold 30, 결과 1이다.

### 왜 필요한가?

### 직접 확인

## Q47. 예상 결과와 실제 결과는 왜 따로 기록하는가?

### 왜 필요한가?

### 왜 필요한가?
