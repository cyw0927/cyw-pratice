### 왜 필요한가?
입력값과 계산 결과의 정상 범위를 알아야 테스트할 때 이상한 값을 빠르게 발견할 수 있다.

### 헷갈리기 쉬운 점
256단계이지만 최댓값은 256이 아니라 255다.

### 직접 해보기

```python
print(2**8)
print(2**8-1)
```

## Q05. 검은색과 흰색의 RGB와 밝기는 각각 얼마인가?

### 왜 필요한가?
최솟값과 최댓값을 먼저 확인하면 계산식이나 자료형 오류를 빠르게 찾을 수 있다.

### 헷갈리기 쉬운 점
흰색 RGB의 합은 765지만 밝기는 평균인 255다.

### 직접 해보기
검은색과 흰색을 직접 계산한다.

```python
black=(0,0,0)
white=(255,255,255)
print(sum(black)//3, sum(white)//3)
```

## Q06. 왜 RGB 세 값을 밝기 하나로 바꾸는가?

### 답
흑백 분류에서는 색의 종류보다 얼마나 밝고 어두운지를 하나의 값으로 비교하는 것이 편하기 때문이다.

### 상세 설명
R,G,B 세 숫자를 그대로 비교하려고 하면 픽셀 하나마다 세 값을 동시에 판단해야 한다. 반면 밝기라는 대표값 하나를 만들면 모든 픽셀을 동일한 threshold와 비교할 수 있다. 즉 색 정보 전체를 유지하려는 것이 아니라, 이 문제에 필요한 밝고 어두운 정도만 뽑아내는 과정이다.

### 왜 필요한가?
0과 1로 분류하려면 모든 픽셀을 같은 기준으로 비교할 수 있는 대표 숫자가 필요하다.

### 헷갈리기 쉬운 점
서로 다른 색도 같은 밝기 값을 가질 수 있다.

### 직접 해보기
빨강 `(255,0,0)`과 회색 `(85,85,85)`의 밝기를 비교한다.

```python
for rgb in [(255,0,0),(85,85,85)]:
    print(rgb, sum(rgb)//3)
```

## Q07. 이 문제에서 밝기는 정확히 어떻게 계산하는가?

### 답
`brightness = (R + G + B) // 3`으로 계산한다.

### 상세 설명
세 색 성분을 모두 더한 뒤 3으로 나눈 정수 평균을 밝기로 사용한다. 실제 영상 처리에서는 사람 눈의 민감도를 반영한 가중치 공식을 쓰기도 하지만, 이 과제에서는 문제에 주어진 단순 평균 공식을 그대로 사용해야 한다. 문제에서 정의한 공식을 바꾸면 예제와 다른 답이 나온다.

### 왜 필요한가?
이 값이 threshold 계산의 재료가 되므로 밝기 공식이 틀리면 이후 모든 결과가 함께 틀어진다.

### 헷갈리기 쉬운 점
실제 영상 처리 공식과 과제 공식은 다를 수 있다.

### 직접 해보기
`(120,120,120)`과 `(255,0,0)`을 손으로 계산한 뒤 코드와 비교한다.

```python
for r,g,b in [(120,120,120),(255,0,0)]:
    print((r+g+b)//3)
```

## Q08. 왜 `/`가 아니라 `//`를 사용하는가?

### 답
소수점 아래를 버린 정수 결과가 필요하기 때문이다.

### 상세 설명
Python에서 `/`는 실수 나눗셈이고 `//`는 몫을 구하는 정수 나눗셈이다. 예를 들어 `7/3`은 약 2.333이고 `7//3`은 2다. 이 문제는 밝기와 threshold를 모두 정수로 계산하므로 `//`를 사용해야 예제와 같은 값이 나온다.

### 왜 필요한가?
연산자 하나 차이로 자료형과 threshold 값이 달라져 최종 분류가 바뀔 수 있다.

### 헷갈리기 쉬운 점
`//`는 반올림이 아니라 아래쪽 정수 몫을 얻는 연산이다.

### 직접 해보기
`7/3`, `7//3`의 결과와 자료형을 비교한다.

```python
print(7/3, type(7/3))
print(7//3, type(7//3))
```

## Q10. 밝기 계산을 함수로 분리하는 이유는 무엇인가?

### 답
같은 계산을 재사용하고 코드의 역할을 명확하게 만들기 위해서다.

### 상세 설명
`calc_brightness()`가 픽셀 하나의 밝기 계산만 담당하도록 만들면 메인 코드가 훨씬 읽기 쉬워진다. 또한 함수만 따로 여러 RGB 값으로 테스트할 수 있어 디버깅하기도 편하다. 나중에 밝기 공식이 바뀌더라도 한 함수만 수정하면 된다.

### 왜 필요한가?
기능을 잘게 나누면 읽기, 테스트, 수정, 설명이 모두 쉬워진다.

### 헷갈리기 쉬운 점
함수 분리는 코드를 길게 만드는 것이 아니라 역할을 분명하게 만드는 작업이다.

### 직접 해보기
검정, 흰색, 빨강을 함수에 넣어 본다.

```python
def calc_brightness(r,g,b):
    return (r+g+b)//3
for rgb in [(0,0,0),(255,255,255),(255,0,0)]:
    print(rgb, calc_brightness(*rgb))
```

## Q11. H와 W는 각각 무엇을 뜻하는가?

### 답
H는 이미지 높이, W는 이미지 너비다.

### 상세 설명
H는 행의 개수이고 W는 한 행에 들어 있는 픽셀의 개수다. 입력이 `3 3`이면 3행 3열 이미지이므로 총 9개의 픽셀이 있다. 이 두 값은 입력을 몇 줄 받아야 하는지와 전체 픽셀 수를 계산할 때 모두 사용된다.

### 왜 필요한가?
이미지 구조를 정확히 이해해야 반복문 횟수와 평균 계산이 맞아진다.

### 헷갈리기 쉬운 점
H는 세로 행 수, W는 가로 열 수다.

### 직접 해보기
`H=3, W=4`에서 총 픽셀 수를 계산한다.

```python
H,W=3,4
print('행:',H,'열:',W,'총 픽셀:',H*W)
```

## Q12. 전체 픽셀 수는 왜 `H * W`인가?

### 답
H개의 행마다 W개의 픽셀이 있기 때문이다.

### 상세 설명
이미지를 표처럼 생각하면 이해하기 쉽다. 3행 4열 표에는 한 행에 4칸씩 총 3줄이 있으므로 12칸이다. 이미지도 같은 구조라 전체 픽셀 수는 높이와 너비의 곱이다.

### 왜 필요한가?
threshold는 전체 밝기 합을 픽셀 수로 나눈 평균이므로 정확한 개수가 필요하다.

### 헷갈리기 쉬운 점
H+W가 아니라 H×W다.

### 직접 해보기
2행 5열과 10행 10열의 픽셀 수를 계산한다.

```python
for H,W in [(2,5),(10,10)]:
    print(H,W,'->',H*W)
```

## Q13. 첫 줄 `3 3`은 어떻게 두 변수에 저장하는가?

### 답
공백 기준으로 나눈 뒤 정수로 바꿔 `height, width`에 저장한다.

### 상세 설명
`input().split()`은 입력 문자열을 공백 기준으로 나눠 두 문자열을 만든다. `map(int, ...)`는 그 문자열들에 `int()`를 적용해 정수로 바꾼다. 마지막으로 언패킹을 사용해 첫 값을 height, 두 번째 값을 width에 저장한다.

### 왜 필요한가?
문자열 상태로는 반복 횟수나 계산에 안전하게 사용할 수 없기 때문에 정수 변환이 필요하다.

### 헷갈리기 쉬운 점
`split()`까지만 하면 여전히 문자열이다.

### 직접 해보기
`'3 3'` 문자열을 실제 input처럼 처리한다.

```python
line='3 3'
height,width=map(int,line.split())
print(height,width,type(height),type(width))
```

## Q15. RGB 문자열 `255,0,0`은 어떻게 세 값으로 나누는가?

### 답
`split(',')`으로 쉼표 기준 세 조각으로 나눈다.

### 상세 설명

새 행을 처리할 때 이전 행에서 사용한 리스트를 그대로 계속 쓰면 값이 한 리스트에 섞여 버린다. 그래서 각 행이 시작될 때 빈 리스트를 만들고, 그 행의 밝기를 모두 넣은 뒤 전체 `brightness_map`에 추가한다.

### 왜 필요한가?

### 왜 필요한가?

## Q45. 문자열 덧붙이기 대신 `join()`을 사용할 수도 있는가?

### 답
가능하다. 각 픽셀 결과를 모은 뒤 `''.join(...)`으로 한 행 문자열을 만들 수 있다.

### 상세 설명
현재 코드는 `result_row += '1'`처럼 문자열을 하나씩 이어 붙인다. 다른 방식은 `'1'`과 `'0'`을 리스트에 모은 뒤 마지막에 join으로 합치는 것이다. 데이터가 매우 많다면 join 방식이 더 효율적일 수 있지만, 이 문제의 최대 크기에서는 둘 다 충분히 동작한다. 중요한 것은 두 방식 모두 최종 결과가 똑같아야 한다는 점이다.

### 왜 필요한가?
같은 출력을 만드는 여러 문자열 처리 방식을 비교할 수 있다.

### 헷갈리기 쉬운 점
`join()`은 문자열 원소를 합치는 기능이므로 정수 리스트라면 먼저 문자열로 바꿔야 한다.

### 직접 해보기
같은 밝기 행을 두 방식으로 변환한다.

```python
row=[120,60,102]
t=102
a=''
for b in row:
    a += '1' if b>=t else '0'
b=''.join('1' if x>=t else '0' for x in row)
print(a,b,a==b)
```

## Q51. 이 문제의 시간 복잡도는 어떻게 생각할 수 있는가?

### 답
모든 픽셀을 몇 번 순회하므로 전체적으로 O(H×W)라고 볼 수 있다.

### 상세 설명
픽셀 수를 N=H×W라고 두면 밝기 계산에서 N개를 보고, threshold 합계에서 다시 N개를 보고, 흑백 변환에서도 N개를 본다. 총 3N 정도의 작업이지만 Big-O에서는 상수 3을 무시하므로 O(N), 즉 O(H×W)로 정리한다. 이미지가 두 배 많은 픽셀을 가지면 처리할 값도 대략 두 배 늘어난다. 반복문이 세 번 있다고 O(N³)이 되는 것이 아니라 각 순회가 순차적으로 이어진다는 점이 중요하다.

### 왜 필요한가?
입력 크기가 커질 때 실행 시간이 어떻게 증가하는지 설명할 수 있다.

### 헷갈리기 쉬운 점
서로 겹쳐 중첩된 세 반복이 아니라 전체 픽셀 순회를 세 번 하는 것이므로 3N이다.

### 직접 해보기
서로 다른 H,W에서 픽셀 수를 출력한다.

```python
for H,W in [(10,10),(100,100),(1000,1000)]:
    print(H,W,'픽셀 수=',H*W)
```

## Q56. RGB 값이 범위를 벗어나는지 검사하려면 어떻게 할 수 있는가?

### 답
세 값 모두 `0 <= value <= 255`인지 확인하면 된다.

### 상세 설명
문제에서는 정상 범위의 입력이 주어진다고 가정하지만 실제 프로그램에서는 입력 검증이 필요할 수 있다. `all()`과 조건식을 사용하면 세 성분이 모두 범위 안에 있는지 간단히 검사할 수 있다. 예를 들어 `(300,0,0)`은 정수 변환 자체는 성공하지만 RGB 정상 범위는 벗어난다. 즉 자료형 검증과 값 범위 검증은 서로 다른 문제다.

### 왜 필요한가?
문제 조건을 코드로 표현하고 데이터 검증 개념을 이해하는 연습이 된다.

### 헷갈리기 쉬운 점
정수라고 해서 반드시 올바른 RGB 값인 것은 아니다.

### 직접 해보기
정상 RGB와 범위를 벗어난 RGB를 검사한다.

```python
for rgb in [(255,0,0),(300,0,0),(-1,10,20)]:
    valid=all(0<=v<=255 for v in rgb)
    print(rgb,valid)
```

## Q59. threshold가 0이 될 수 있는가?

### 상세 설명

```python
values=[0,0,0,0]
t=sum(values)//len(values)
print(t,[1 if v>=t else 0 for v in values])
```

## Q75. 75개 질문을 모두 학습한 뒤 문제를 다시 풀면 어떤 순서로 접근하는 것이 좋은가?

### 답
입력 구조 확인 → 한 픽셀 밝기 계산 → 밝기 맵 구성 → 전체 평균 threshold 계산 → 0/1 변환 → 출력 형식 확인 → 테스트 순서로 접근하면 된다.

### 상세 설명
처음부터 전체 코드를 한 번에 작성하려고 하면 입력 파싱, 평균 계산, 조건문, 출력 중 어디에서 틀렸는지 찾기 어렵다. 반대로 각 단계를 작은 함수나 작은 예제로 확인하면서 연결하면 오류가 생겨도 범위를 좁힐 수 있다. 먼저 RGB 한 픽셀이 제대로 숫자로 바뀌는지 확인하고, 그다음 한 행, 전체 이미지 순서로 범위를 넓힌다. 마지막에는 문제 예제뿐 아니라 1×1, 균일 이미지, threshold와 같은 값 같은 경계 테스트까지 실행해 기대값과 실제값이 같은지 확인한다. 이렇게 하면 코드를 외워 쓰는 것이 아니라 알고리즘의 흐름을 이해한 상태에서 다시 구현할 수 있다.

### 왜 필요한가?
75개의 세부 질문을 하나의 문제 해결 절차로 다시 합치는 마지막 정리 단계다.

### 헷갈리기 쉬운 점
질문 75개를 따로 외우는 것이 목적이 아니라 하나의 전체 흐름으로 연결하는 것이 목적이다.

### 직접 해보기
문제 전체 흐름을 함수 세 개로 구성한 작은 완성 예제를 실행한다.

```python
def calc_brightness(r,g,b): return (r+g+b)//3
def calc_threshold(m): return sum(sum(row) for row in m)//sum(len(row) for row in m)
def to_bw(m,t): return [''.join('1' if b>=t else '0' for b in row) for row in m]
rgb_rows=[[(255,255,255),(0,0,0)],[(120,120,120),(60,60,60)]]
m=[[calc_brightness(*p) for p in row] for row in rgb_rows]
t=calc_threshold(m)
print('brightness_map=',m)
print('threshold=',t)
print('result=',to_bw(m,t))
```
